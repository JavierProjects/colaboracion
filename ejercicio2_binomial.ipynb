# Ejercicio 2. Distribución Binomial

**Sexualidad.** Con base en la Encuesta Nacional de Juventud 2005, el **90%** de los adolescentes entre 15 y 19 años afirmó que tuvo su primera relación sexual antes de los 18 años. Si en un grupo hay **20 adolescentes** entre 15 y 19 años:

- a) ¿Cuántos esperarías que contestaran que sí? (Valor esperado)
- b) Probabilidad de que **exactamente 12** digan que sí.
- c) Probabilidad de que **más de 17** digan que sí.
- d) Discusión sobre métodos anticonceptivos, ETS y consecuencias.

## Variable aleatoria

$$X = \text{número de adolescentes (de 15 a 19 años) que tuvieron su primera relación sexual antes de los 18, en un grupo de 20.}$$

El experimento cumple los axiomas de un experimento binomial:

- Número fijo de ensayos: $n = 20$.
- Solo dos resultados: sí o no.
- Probabilidad de éxito constante: $p = 0.90$.
- Ensayos independientes.

## Distribución Binomial

Una variable aleatoria $X$ sigue una distribución Binomial con parámetros $n \in \mathbb{N}$ y $p \in [0,1]$ si su función de masa de probabilidad es:

$$P(X = x) = \binom{n}{x}\,p^x\,(1-p)^{n-x}, \quad x = 0, 1, \ldots, n$$

En nuestro caso:

$$X \sim \text{Binomial}(n = 20,\; p = 0.90)$$

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom
from IPython.display import HTML, display

print("Librerías cargadas correctamente")
```

## Parámetros del modelo

| Parámetro | Valor |
|-----------|-------|
| $n$ | 20 |
| $p$ | 0.90 |
| $q = 1 - p$ | 0.10 |
| $E(X) = n \cdot p$ | 18 |
| $Var(X) = n \cdot p \cdot q$ | 1.80 |

```python
n, p = 20, 0.90
q = 1 - p

print(f"n            = {n}")
print(f"p            = {p}")
print(f"q = 1 - p    = {q}")
print(f"E(X) = n*p   = {n*p}")
print(f"Var(X) = npq = {n*p*q}")
print(f"DE(X)        = {np.sqrt(n*p*q):.4f}")
```

## Tabla de distribución completa

```python
x_vals = np.arange(0, 21)
probs = binom.pmf(x_vals, n, p)

filas = ""
for xi, pi in zip(x_vals, probs):
    bg = "#f4f6f7" if xi % 2 == 0 else "#ffffff"
    filas += f"""
    <tr style="background-color: {bg};">
      <td style="text-align:center; padding:6px; border:1px solid #bbb;">{xi}</td>
      <td style="text-align:center; padding:6px; border:1px solid #bbb;">{pi:.6f}</td>
      <td style="text-align:center; padding:6px; border:1px solid #bbb;">{pi*100:.4f}%</td>
    </tr>"""

html_tabla = f"""
<table style="border-collapse:collapse; margin:auto; font-family:Arial; width:50%;">
  <thead>
    <tr style="background-color:#3498db; color:white;">
      <th style="padding:8px; border:1px solid #bbb;">x</th>
      <th style="padding:8px; border:1px solid #bbb;">P(X = x)</th>
      <th style="padding:8px; border:1px solid #bbb;">%</th>
    </tr>
  </thead>
  <tbody>{filas}
  </tbody>
</table>
<p style="text-align:center; font-family:Arial; color:#555;">
  <i>Tabla 1. Distribucion Binomial(n = 20, p = 0.90)</i>
</p>
"""
display(HTML(html_tabla))
```

```python
fig, ax = plt.subplots(figsize=(11, 5))
bars = ax.bar(x_vals, probs, color="#a9cce3", edgecolor="#2c3e50", linewidth=1.0)

# Resaltar la moda (x = 18)
for b, xi in zip(bars, x_vals):
    if xi == 18:
        b.set_color("#e74c3c")

for b, pi in zip(bars, probs):
    if pi > 0.005:
        ax.text(b.get_x() + b.get_width()/2, b.get_height() + 0.005,
                f"{pi:.4f}", ha="center", fontsize=8)

ax.set_title("Distribucion Binomial(n = 20, p = 0.90)",
             fontsize=13, fontweight="bold")
ax.set_xlabel("x = adolescentes con primera relacion sexual antes de los 18", fontsize=11)
ax.set_ylabel("P(X = x)", fontsize=11)
ax.set_xticks(x_vals)
ax.set_ylim(0, max(probs)*1.15)
ax.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

## Análisis del sesgo

Como $p = 0.90 > 0.5$, el coeficiente de asimetría es:

$$\gamma_1 = \frac{q - p}{\sqrt{npq}} = \frac{0.10 - 0.90}{\sqrt{20 \cdot 0.90 \cdot 0.10}} = \frac{-0.80}{1.3416} = -0.5963$$

**Sesgo negativo (asimétrica hacia la izquierda); mayor probabilidad puntual en $x = 18$ con $P(X=18) = 0.2852$.**

## a) Valor esperado

**Fórmula:**
$$E(X) = n \cdot p$$

**Sustitución:**
$$E(X) = 20 \cdot 0.90 = 18$$

**Respuesta: 18 adolescentes.**

```python
esperanza = n * p
print(f"E(X) = n·p = 20·0.90 = {esperanza}")
print(f"Interpretación: se esperaría que {esperanza:.0f} adolescentes "
      f"respondan afirmativamente.")
```

## b) Probabilidad de exactamente 12

**Fórmula:**
$$P(X = 12) = \binom{20}{12}\,(0.9)^{12}\,(0.1)^{8}$$

**Sustitución:**
$$P(X = 12) = 125970 \times 0.282430 \times 10^{-8} = 0.0003558$$

**Respuesta: 0.0003558 aprox. o 0.0356%.**

```python
p_12 = binom.pmf(12, n, p)
print(f"C(20,12)  = 125970")
print(f"(0.90)^12 = {0.90**12:.6f}")
print(f"(0.10)^8  = {0.10**8}")
print(f"P(X = 12) = {p_12:.6f}")
print(f"            = {p_12*100:.4f}%")
```

## c) Probabilidad de más de 17

"Más de 17" equivale a la unión disjunta $\{X = 18\} \cup \{X = 19\} \cup \{X = 20\}$:

$$P(X > 17) = P(X = 18) + P(X = 19) + P(X = 20)$$

**Cálculo de cada término:**

$$P(X = 18) = \binom{20}{18}\,(0.9)^{18}\,(0.1)^{2} = 190 \times 0.150095 \times 0.01 = 0.285180$$

$$P(X = 19) = \binom{20}{19}\,(0.9)^{19}\,(0.1)^{1} = 20 \times 0.135145 \times 0.10 = 0.270170$$

$$P(X = 20) = \binom{20}{20}\,(0.9)^{20}\,(0.1)^{0} = 1 \times 0.121577 \times 1 = 0.121577$$

**Suma total:**

$$P(X > 17) = 0.285180 + 0.270170 + 0.121577 = 0.676927$$

**Respuesta: 0.6769 aprox. o 67.69%.**

```python
p_18 = binom.pmf(18, n, p)
p_19 = binom.pmf(19, n, p)
p_20 = binom.pmf(20, n, p)
p_mayor_17 = p_18 + p_19 + p_20

print(f"P(X = 18)  = {p_18:.6f}")
print(f"P(X = 19)  = {p_19:.6f}")
print(f"P(X = 20)  = {p_20:.6f}")
print(f"P(X > 17)  = {p_mayor_17:.6f}  =  {p_mayor_17*100:.4f}%")
```
