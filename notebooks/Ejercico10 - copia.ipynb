<span style="color: blue;">Sexualidad</span>. Con base en la Encuesta Nacional de Juventud 2005, 90% de los adolescentes entre 15 y 19 años afirmó que tuvo su primera relación sexual antes de los 18 años. Si en un grupo hay 20 adolescentes entre 15 y 19 años, entonces:

a) ¿Cuántos esperarías que contestaran que tuvieron su primera relación sexual antes de los 18 años?  
b)¿Cuál es la probabilidad de que 12 de ellos digan que tuvieron su primera relación sexual antes de los 18 años? 

c) ¿Cuál es la probabilidad de que más de 17 digan que tuvieron su primera relación sexual antes de los 18 años?  

Sea:

$$X=\mbox{numero de adolescentes que tuvo su primera relación sexual}$$

Como hay 20 adolescentes y cada uno tiene probabilidad p=0.9 que tuvo su primera relación sexual:

$$X∼\mbox{Binomial}(n=20,p=0.9)$$

## Distribución de probabilidad

La función de probabilidad es:

$$P(X=x)=\binom{20}{x}(0.9)^x(0.10)^{10-x} \quad \mbox{con} \quad x=0,1,2,\dots,10$$

```python
import matplotlib.pyplot as plt
from scipy.stats import binom

# Parámetros
n = 20
p = 0.9

# Valores posibles de X
x = list(range(n + 1))

# Distribución binomial
probabilidades = binom.pmf(x, n, p)

# Mostrar tabla en consola
#for k, prob in zip(x, probabilidades):
#    print(f"P(X = {k}) = {prob:.6f}")

# ==========================
# Generar tabla HTML
# ==========================

html = """
<table style="border-collapse: collapse; width: 420px; text-align: center;">
  <thead>
    <tr>
      <th style="border: 1px solid #000; padding: 10px; width: 160px;">x</th>
      <th style="border: 1px solid #000; padding: 10px; width: 260px;">P(X=x)</th>
    </tr>
  </thead>
  <tbody>"""

for k, prob in zip(x, probabilidades):
    html += f"""
    <tr>
      <td style="border: 1px solid #000; padding: 8px;">{k}</td>
      <td style="border: 1px solid #000; padding: 8px;">{prob:.6f}</td>
    </tr>"""

html += """
  </tbody>
</table>
"""

# Mostrar el HTML en la consola
print("\nCódigo HTML:\n")
print(html)
```

```python
# ==========================
# Gráfica
# ==========================

plt.figure(figsize=(8, 5))
plt.bar(x, probabilidades, width=1, edgecolor="w")

plt.title("adolescentes entre 15 y 19 años afirmó que tuvo su primera relación sexua")
plt.xlabel("Número de adolescentes que tuvo su primera relación sexual")
plt.ylabel("Probabilidad")
plt.xticks(x)
plt.grid(axis="y", linestyle="--", alpha=0.6)

plt.show()
```

##  Tipo de sesgo

La distribución presenta sesgo a la izquierda (Sesgo Negativo).

Esto ocurre porque $p=0.9$, es decir, la probabilidad de éxito es muy alta que $0.5$. La mayor parte de la probabilidad se concentra en valores maximos de $X$, principalmente en los valores más bajos alrededor de 13, 14 y 15, mientras que la gran mayoría de las barras altas se concentran a la derecha.

## a) Número esperado de su primera relación sexual antes de los 18 años 

Para una distribución binomial:

$$E(X)=np$$

Por lo tanto tenemos:

$$
\begin{align*}
E(X) &= np \\
     &= 20(0.9) \\
     &= 18
\end{align*}
$$

Por lo tanto, se esperarían:

$$18$$

adolescentes contesten de forma afirmativa.

## c) Probabilidad de que de que 12 de ellos digan que tuvieron su primera relación sexual antes de los 18 años

$$
\begin{align*}
P(X=5) &= \binom{20}{12}(0.9)^12(0.10)^{20-12} \\
       &= \binom{20}{12}(0.9)^12(0.10)^8 \\
       &=  125.970(0.28242)(0.00000001) \\
       &= 0.000356
\end{align*}
$$

Por lo tanto:

$$P(X=12)\approx 0.000356$$

Es decir, aproximadamente:

$$0.036\%$$

## d) Probabilidad de que más de 17 digan que tuvieron su primera relación sexual antes de los 18 años

“Al menos 12” significa:

$$P(X > 12)$$

Conviene usar el complemento:

$$P(X > 12)$$

Es decir:

$$P(X > 17) = P(X = 18) + P(X = 19) + P(X = 20)$$

Por lo tanto: 
$$
\begin{aligned}
P(X > 17) &= \left[\binom{20}{18}(0.90)^{18}(0.10)^2\right] + \left[\binom{20}{19}(0.90)^{19}(0.10)^1\right] + \left[\binom{20}{20}(0.90)^{20}(0.10)^0\right] \\
&= [190 \cdot 0.15009 \cdot 0.01] + [20 \cdot 0.13509 \cdot 0.10] + [1 \cdot 0.12158 \cdot 1] \\
&= 0.28518 + 0.27017 + 0.12158 \\
&= 0.67693
\end{aligned}
$$

Por lo tanto:

$$P(X > 17) \approx 0.67693$$

Es decir, aproximadamente: 
$$67.69\% $$
