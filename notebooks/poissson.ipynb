## a) Determina la probabilidad de que en un día coloque entre 100 y 110 piezas.

**Datos:**
* Rango original: $100 \le X \le 110$
* Con corrección por continuidad: $99.5 \le X \le 110.5$

**Fórmula:**
$$P(99.5 \le X \le 110.5) = P\left(\frac{99.5 - \mu}{\sigma} \le Z \le \frac{110.5 - \mu}{\sigma}\right)$$

**Sustitución:**
$$P(99.5 \le X \le 110.5) = P\left(\frac{99.5 - 100}{10} \le Z \le \frac{110.5 - 100}{10}\right)$$

**Desarrollo:**
$$P(-0.05 \le Z \le 1.05) = P(Z \le 1.05) - P(Z \le -0.05)$$

Consultando la tabla de la distribución normal estándar:
* Para $Z = 1.05 = 0.8531$
* Para $Z = -0.05 = 0.4801$


para Z = la tabla poniendo el 0.05 a positivo restandole 1 = 0.5199-1 = $0.4801$

$$0.8531 - 0.4801$$

**Resultado:**
$$P(100 \le X \le 110) = 0.3730$$
