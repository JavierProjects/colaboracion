# Ejercicio 13
El Mar de Cortés, en México, es la zona donde se localiza el mayor número de granjas dedicadas al cultivo de ostras perleras. Según los productores, **3.6 de cada cien ostras** pueden llegar a tener un buen valor comercial. Si tenemos **150 ostras**:

a) Determina la probabilidad de que 12 de ellas tengan perlas de buen valor comercial.  
b) Encuentra la probabilidad de que cuando mucho dos tengan perlas de buen valor comercial.

Sea:

$$X=\text{número de ostras que tienen perlas de buen valor comercial}$$

Se tiene un número fijo de ostras:

$$n=150$$

La probabilidad de que una ostra tenga una perla de buen valor comercial es:

$$p=\frac{3.6}{100}=0.036$$

La probabilidad de que no tenga una perla de buen valor comercial es:

$$q=1-p=1-0.036=0.964$$

Por lo tanto:

$$X\sim\operatorname{Binomial}(n=150,p=0.036)$$

La función de probabilidad binomial es:

$$P(X=x)=\binom{n}{x}p^xq^{n-x}$$

Para este problema:

$$P(X=x)=\binom{150}{x}(0.036)^x(0.964)^{150-x}$$

## a) Probabilidad de que exactamente 12 tengan perlas de buen valor comercial

Se desea calcular:

$$P(X=12)$$

Sustituimos los valores:

$$
\begin{align*}
P(X=12)
    &= \binom{150}{12}(0.036)^{12}(0.964)^{150-12} \\
    &= \binom{150}{12}(0.036)^{12}(0.964)^{138} \\
    &\approx 0.005186
\end{align*}
$$

En porcentaje:

$$
\boxed{P(X=12)\approx 0.5186\%}
$$

## b) Probabilidad de que cuando mucho dos tengan perlas de buen valor comercial

“Cuando mucho dos” significa:

$$X\leq 2$$

Por lo tanto:

$$P(X\leq 2)=P(X=0)+P(X=1)+P(X=2)$$

Sustituimos:

$$
\begin{align*}
P(X\leq 2)
    &= \binom{150}{0}(0.036)^0(0.964)^{150} \\
    &\quad +\binom{150}{1}(0.036)^1(0.964)^{149} \\
    &\quad +\binom{150}{2}(0.036)^2(0.964)^{148} \\
    &\approx 0.004088+0.022902+0.063717 \\
    &\approx 0.090707
\end{align*}
$$

En porcentaje:

$$
\boxed{P(X\leq 2)\approx 9.0707\%}
$$

## Resultados

$$
\boxed{\text{a) }P(X=12)\approx 0.005186}
$$

$$
\boxed{\text{b) }P(X\leq 2)\approx 0.090707}
$$
