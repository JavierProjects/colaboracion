<span style="color: blue;">Salud</span>. 14 De acuerdo con la Encuesta de Consumo de Alcohol, Tabaco y otras Drogas en la Ciudad de México (2009), **25.3% de las personas consumen marihuana, drogas inhalables o cocaína**. Si tenemos un grupo de 25 personas:

a) ¿Cuál es la probabilidad de que 10 de ellas consuman alguna de estas drogas?  
b) ¿Cuál es la probabilidad de que al menos 20 de ellas consuman alguna de estas drogas?  
c) De las siguientes 50 personas, ¿cuántas esperarías que consumieran alguna de estas drogas?  
d) Actualmente, el problema de consumo de drogas se presenta en todos los estratos sociales. Por lo general se utilizan para “estar en ambiente”, o bien, para “evadir la realidad”, entre otras muchas razones. ¿Sabes las consecuencias que el consumo de drogas trae consigo? ¿Crees que las campañas para vivir sin drogas han sido eficaces? Sea:

$$X = \text{número de personas que consumen alguna de estas drogas}$$

Como hay un número fijo de ensayos independientes ($n=25$), cada persona tiene solo dos posibilidades (consume o no consume) y la probabilidad de éxito es constante ($p=0.253$), utilizamos una **distribución binomial**:

$$X \sim \operatorname{Binomial}(n = 25, p = 0.253)$$

La función de probabilidad binomial es:

$$P(X = k) = \binom{n}{k} p^k (1-p)^{n-k}$$

## a) Distribución de probabilidad y gráfica correspondiente


### Probabilidad de que exactamente 10 consuman de estas drogas

Calculamos $P(X = 10)$:

$$
\begin{align*}
P(X = 10) &= \binom{25}{10} (0.253)^{10} (1 - 0.253)^{25 - 10} \\
         &= (3,268,760)(0.0000010168)(0.015243) \\
         &= 0.050664
\end{align*}
$$

Por lo tanto:

$$P(X = 10) \approx 0.0507$$

Es decir, aproximadamente:
$$5.07\%.$$

## b) Probabilidad de que al menos 20 de ellas consuman alguna de estas drogas



$$P(X \ge 20)$$

Sumamos los valores individuales desde $k = 20$ hasta $n = 25$:

$$
\begin{align*}
P(X \ge 20) &= P(X=20) + P(X=21) + P(X=22) + P(X=23) + P(X=24) + P(X=25) \\
            &= \sum_{k=20}^{25} \binom{25}{k} (0.253)^k (0.747)^{25-k} \\
            &= 0.0000000003 + 0.0000000000 + \dots \\
            &\approx 0.000000
\end{align*}
$$

Debido a que la probabilidad de éxito es del $25.3\%$, la probabilidad de encontrar un número tan alto como 20 o más consumidores en un grupo tan pequeño de 25 personas de forma aleatoria es extremadamente baja (prácticamente cero, específicamente $3.29 \times 10^{-10}$).

Por lo tanto:

$$P(X \ge 20) \approx 0.0000$$

Es decir, aproximadamente:
$$0.00\%.$$

## c) Número esperado de consumidores en las siguientes 50 personas
Fórmula:

$$E(X) = n \cdot p$$

 $n = 50$ personas:

$$
\begin{align*}
E(X) &= 50 \times 0.253 \\
     &= 12.65
\end{align*}
$$

Por lo tanto, se esperaría que aproximadamente **12.65 personas** (o entre 12 y 13 personas) de las siguientes 50 consuman alguna de estas drogas.

<span style="color: blue;">Inseguridad</span>. 18 De acuerdo con las estadísticas de criminalidad, en promedio se roba un auto cada tres minutos. Determina:

a) La distribución de probabilidad y la gráfica correspondiente. Describe de manera breve la distribución.
b) La probabilidad de que en los siguientes tres minutos se roben dos autos.
c) La probabilidad de que en los siguientes tres minutos se roben dos autos o menos.
d) La probabilidad de que se roben seis autos en los siguientes 15 minutos.
e) Uno de los factores que afectan los costos de los seguros de automóviles es el número de robos (siniestros) que se cometen. ¿Qué se podría hacer para evitar el robo de autos?.

Sea:

$$X = \text{número de autos robados en un intervalo de tiempo}$$

La función de probabilidad de Poisson es:

$$P(X = k) = \frac{e^{-\lambda} \lambda^k}{k!}$$

## a) Distribución de probabilidad y gráfica para un intervalo de 3 minutos

El intervalo de tiempo de interés es de 3 minutos. Dado que en promedio se roba un auto cada 3 minutos, el parámetro de la distribución es:

$$\lambda = 1$$

Por lo tanto, la variable aleatoria sigue una distribución:

$$X \sim \operatorname{Poisson}(1)$$

La distribución de probabilidad de los primeros valores de $k$ es:

- $P(X=0) = \frac{e^{-1} 1^0}{0!} = e^{-1} \approx 0.367879$
- $P(X=1) = \frac{e^{-1} 1^1}{1!} = e^{-1} \approx 0.367879$
- $P(X=2) = \frac{e^{-1} 1^2}{2!} = \frac{e^{-1}}{2} \approx 0.183940$
- $P(X=3) = \frac{e^{-1} 1^3}{3!} = \frac{e^{-1}}{6} \approx 0.061313$
- $P(X=4) = \frac{e^{-1} 1^4}{4!} = \frac{e^{-1}}{24} \approx 0.015328$

**Descripción de la distribución:**
La distribución presenta un **sesgo positivo (a la derecha)**. Esto significa que la mayor concentración de la probabilidad se localiza en los valores bajos (0 y 1 auto robado), y a medida que aumenta el número de autos robados, la probabilidad disminuye rápidamente, extendiendo una "cola" larga hacia la derecha. La tasa promedio es baja ($\lambda = 1$), haciendo que sea muy poco probable observar una cantidad elevada de robos en tan solo 3 minutos.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson

# Parámetro lambda para 3 minutos
lam = 1
k_values = np.arange(0, 8)
probabilities = poisson.pmf(k_values, lam)

# Graficar la distribución
plt.figure(figsize=(8, 5))
plt.bar(k_values, probabilities, color='skyblue', edgecolor='black', alpha=0.7, label='P(X=k)')
plt.step(k_values, probabilities, where='mid', color='blue', linestyle='--', alpha=0.5)
plt.title('Distribución de Probabilidad de Poisson ($\lambda = 1$ para 3 minutos)')
plt.xlabel('Número de autos robados (k)')
plt.ylabel('Probabilidad')
plt.xticks(k_values)
plt.grid(axis='y', linestyle='--', alpha=0.7)
for i, p in enumerate(probabilities):
    plt.text(k_values[i], p + 0.01, f'{p:.4f}', ha='center', fontsize=9)
plt.ylim(0, 0.45)
plt.show()
```

## b) Probabilidad de que en los siguientes tres minutos se roben dos autos

Queremos calcular $P(X = 2)$ con $\lambda = 1$:

$$
\begin{align*}
P(X = 2) &= \frac{e^{-1}(1)^2}{2!} \\
         &= \frac{e^{-1}(1)}{2} \\
         &= \frac{0.367879}{2} \\
         &= 0.183940
\end{align*}
$$

La probabilidad de que se roben exactamente dos autos en los siguientes tres minutos es de aproximadamente **18.39%**.

## c) Probabilidad de que en los siguientes tres minutos se roben dos autos o menos

La expresión "dos autos o menos" significa:

$$P(X \leq 2)$$

 $k = 0$ hasta $k = 2$:

$$
\begin{align*}
P(X \leq 2) &= P(X = 0) + P(X = 1) + P(X = 2) \\
            &= \frac{e^{-1}(1)^0}{0!} + \frac{e^{-1}(1)^1}{1!} + \frac{e^{-1}(1)^2}{2!} \\
            &= 0.367879 + 0.367879 + 0.183940 \\
            &= 0.919698
\end{align*}
$$

La probabilidad de que se roben dos autos o menos en los siguientes tres minutos es de aproximadamente **91.97%**.

## d) Probabilidad de que se roben seis autos en los siguientes 15 minutos

 $\lambda$ de manera proporcional:

$$\lambda_{15} = 1 \times \left(\frac{15}{3}\right) = 5 \text{ autos cada 15 minutos}$$

Definimos una nueva variable aleatoria $Y \sim \operatorname{Poisson}(5)$,  calcular $P(Y = 6)$:

$$
\begin{align*}
P(Y = 6) &= \frac{e^{-5}(5)^6}{6!} \\
         &= \frac{(0.006738)(15625)}{720} \\
         &= \frac{105.2806}{720} \\
         &= 0.146223
\end{align*}
$$

La probabilidad de que se roben exactamente seis autos en un lapso de 15 minutos es de aproximadamente **14.62%**.
