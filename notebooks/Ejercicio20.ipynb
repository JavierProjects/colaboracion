<span style="color: blue;">Deportes</span>. En un gimnasio sólo para mujeres se realizan ejercicios aeróbicos y ejercicios en aparatos. El circuito está constituido por nueve aparatos tonificadores y ocho estaciones para los ejercicios aeróbicos. En un día cualquiera, en hora pico (entre las 19 y 20 horas), llegan en promedio siete mujeres. En ese horario:  

a) ¿Cuál es la probabilidad de que lleguen menos de cinco mujeres?   
b) ¿Cuál es la probabilidad de que estén ocupados todos los aparatos?  
c) Y en un día extraordinario, ¿cuál es la probabilidad de que tanto los aparatos como las estaciones estén ocupados?    

Sea:

$$X=\text{número de mujeres que llegan al gimnasio entre las 19 y 20 horas}$$

Como se conoce el número promedio de llegada de personas al gimnasio cuenta con un intervalo de tiempo continuo (1 hora) con una tasa promedio conocida, utilizamos una **distribución de Poisson** con parámetro:


$$\lambda=7$$

Por lo tanto:

$$X\sim\operatorname{Poisson}(7)$$

La función de probabilidad de Poisson es:


$$P(X=x)=\frac{e^{-\lambda}\lambda^x}{x!}, \quad \mbox{con} \quad x=0,1,2,\ldots$$

Para este problema:


$$P(X=x)=\frac{e^{-7}(7)^x}{x!}$$

##  Distribución de probabilidad y gráfica

La distribución de Poisson puede tomar una cantidad infinita de valores:


$$x=0,1,2,3,\ldots$$

Para construir una tabla y una gráfica manejables, mostraremos los valores de $(x=0)$ a $(x=9)$. En este intervalo se concentra más del $(99.97\%)$ de la probabilidad total.

Tabla de distribucion de probabilidad con valores aproximadados:

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson
from IPython.display import HTML, display

# Parámetro de la distribución
lam = 7

# Valores que se mostrarán en la tabla y en la gráfica
x = np.arange(0, 17)

# Probabilidades P(X = x)
probabilidades = poisson.pmf(x, mu=lam)

# Construcción de la tabla HTML
html = """
<table style="
    border-collapse: collapse;
    width: 520px;
    text-align: center;
    font-family: Arial, sans-serif;
">
    <thead>
        <tr>
            <th style="
                border: 1px solid #000;
                padding: 10px;
                width: 200px;
            ">x</th>
            <th style="
                border: 1px solid #000;
                padding: 10px;
                width: 320px;
            ">P(X = x)</th>
        </tr>
    </thead>
    <tbody>
"""

for valor, probabilidad in zip(x, probabilidades):
    html += f"""
        <tr>
            <td style="border: 1px solid #000; padding: 8px;">
                {valor}
            </td>
            <td style="border: 1px solid #000; padding: 8px;">
                {probabilidad:.6f}
            </td>
        </tr>
    """

html += """
    </tbody>
</table>
"""

display(HTML(html))
```

```python
# Gráfica de la distribución
plt.figure(figsize=(9, 5))
plt.bar(x, probabilidades, width=1, edgecolor="w")

plt.title("Distribución de Poisson: mujeres que llegan al gimnasio entre las 19 y 20 horas")
plt.xlabel("Número de mujeres que llegan al gimnasio")
plt.ylabel("Probabilidad")
plt.xticks(x)
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.tight_layout()

plt.show()

```

### Descripción de la distribución

La mayor concentración de probabilidad se encuentra alrededor de **6 y 7 habitantes por vivienda**.

La distribución presenta **sesgo positivo o sesgo a la derecha**, porque:

- la mayor parte de la probabilidad se concentra en los valores pequeños y cercanos al promedio;
- conforme aumenta el número de habitantes, las probabilidades disminuyen;
- la cola de la distribución se extiende hacia los valores grandes de $(X)$.

El sesgo no es extremadamente pronunciado porque el valor de $(\lambda=7)$ ya permite que la distribución tenga una forma relativamente cercana a una campana, aunque todavía es asimétrica hacia la derecha.

## a) Probabilidad de que lleguen menos de cinco mujeres

La expresión “menos de cinco mujeres” significa:

$$P(X<5)$$

Utilizamos la regla del complemento:

$$
\begin{align*}
P(X<5)
    &= 1-P(X\leq 5) \\
    &= 1-\left[P(X=0)+P(X=1)+P(X=2)+P(X=3)\right] \\
    &= 1-\left[
        \frac{e^{-7}(7)^0}{0!}
        +\frac{e^{-7}(7)^1}{1!}
        +\frac{e^{-7}(7)^2}{2!}
        +\frac{e^{-7}(7)^3}{3!}
        +\frac{e^{-7}(7)^4}{4!}
        \right] \\
    &= 1-\left[
        0.00091 + 0.00638 + 0.02234 + 0.05213 + 0.09123
        \right] \\
    &= 1-0.17299 \\
    &= 0.1730
\end{align*}
$$

Por lo tanto:

$$P(X<5) = 0.1730$$

La probabilidad de que llegen a un gimnasio menos de cinco mujeres es de  aproximadamente $17.30\%$.

## b) Probabilidad de que esten ocupados todos los aparatos 

“Al menos nueve” significa:


$$P(X \ge 9)$$

Conviene usar el complemento:

$$P(X \ge 9) = 1 − P(X < 9)$$
Es decir:

$$P(X \ge 9)=1−[P(X=0)+P(X=1)+P(X=2)+P(X=3)+P(X=4)+P(X=5)+P(X=6)+P(X=7)+P(X=8)]$$

Por lo tanto:

$$
\begin{align*}
P(X \geq 9)
    &= 1 - P(X < 9) \\
    &=1 - \left[P(X=0)+P(X=1)+P(X=2)+P(X=3)+P(X=4)+
    P(X=5)+P(X=6)+P(X=7)+P(X=8)\right] \\
    &= 1 - \left[
        \frac{e^{-7}(7)^0}{0!}
        +\frac{e^{-7}(7)^1}{1!}
        +\frac{e^{-7}(7)^2}{2!}
        +\frac{e^{-7}(7)^3}{3!}
        +\frac{e^{-7}(7)^4}{4!}
        +\frac{e^{-7}(7)^5}{5!}
        +\frac{e^{-7}(7)^6}{6!}
        +\frac{e^{-7}(7)^7}{7!}
        +\frac{e^{-7}(7)^8}{8!}
        \right] \\
    &= 1-\left[
        0.00091 + 0.00638 + 0.02234 + 0.05213 + 0.09123 + 0.12772 + 0.14900 + 0.14900 + 0.13038
        \right] \\
    &= 1- 0.72910 \\
    &= 0.27090
\end{align*}
$$

Por lo tanto:

$$P(X \ge 9) \approx 0.27090$$


Es decir, aproximadamente:

 $$27.09\%$$.


## c) Probabilidad de que tanto los aparatos como las estaciones estén ocupados

Para este inciso sumé la capacidad instalada total del gimnasio, combinando los 9 aparatos tonificadores con las 8 estaciones de aeróbicos, lo que nos da un cupo máximo de $9 + 8 = 17$ espacios disponibles. Para que todo el gimnasio esté lleno, la llegada de mujeres debe ser de 17 o más, lo que representé como:

$$P(X \ge 17)$$

Nuevamente empleé el enfoque del complemento para resolver la operación:

$$P(X \ge 17) = 1 - P(X \le 16)$$

Apoyándome en las tablas de distribución acumulada de Poisson para una media de $\lambda = 7$, identifiqué que la probabilidad de que lleguen 16 personas o menos es sumamente alta debido a la naturaleza de la distribución, alcanzando un valor de $0.9993$.

Efectuando la diferencia:

$$
\begin{aligned}
P(X \geq 17) &= 1 - P(X \leq 16) \\
&= 1 - 0.9993 \\
&= 0.0007
\end{aligned}
$$

Por lo tanto:

$$P(X \ge 17) \approx 0.0007$$

La probabilidad de que el gimnasio se llene por completo en su totalidad es de apenas **0.07%**. Esto resulta coherente estadísticamente, puesto que la llegada de 17 o más personas se aleja considerablemente del promedio habitual de 7.
