Sea:

$$X=\text{número de autos que ingresan a la caseta en un minuto}$$

Estamos contando cuántas veces ocurre un evento (el ingreso de un auto) dentro de un intervalo de tiempo fijo, conociendo únicamente la tasa promedio con la que ocurre. Este es justo el escenario que modela la **distribución de Poisson**, cuyo parámetro $\lambda$ es la tasa promedio de ocurrencia en el intervalo considerado. Aquí:

$$\lambda=75\ \text{autos por minuto}$$

Por lo tanto:

$$X\sim\operatorname{Poisson}(75)$$

La función de probabilidad de Poisson es:

$$P(X=x)=\frac{e^{-\lambda}\lambda^{x}}{x!}, \qquad x=0,1,2,\ldots$$

Para este problema:

$$P(X=x)=\frac{e^{-75}(75)^{x}}{x!}$$

## a) Probabilidad de que ingresen 80 autos

Se pide:

$$P(X=80)=\frac{e^{-75}(75)^{80}}{80!}$$

Calculamos cada parte por separado:

$$e^{-75}\approx 2.678637\times10^{-33}$$

$$(75)^{80}\approx 1.011349\times10^{150}$$

$$80!\approx 7.156946\times10^{118}$$

Multiplicamos primero el numerador:

$$e^{-75}(75)^{80}=(2.678637\times10^{-33})(1.011349\times10^{150})=2.709037\times10^{117}$$

Y dividimos entre $80!$:

$$
\begin{align*}
P(X=80) &= \frac{2.709037\times10^{117}}{7.156946\times10^{118}}\\
        &= 0.037852
\end{align*}
$$

Por lo tanto:

$$P(X=80)\approx 0.037852$$

Es decir, la probabilidad de que en un minuto dado ingresaran exactamente 80 autos es de aproximadamente **3.79%**.

## b) Probabilidad de que ingresen 50 autos

Se pide:

$$P(X=50)=\frac{e^{-75}(75)^{50}}{50!}$$

Calculamos cada parte por separado:

$$e^{-75}\approx 2.678637\times10^{-33}$$

$$(75)^{50}\approx 5.663217\times10^{93}$$

$$50!\approx 3.041409\times10^{64}$$

Multiplicamos primero el numerador:

$$e^{-75}(75)^{50}=(2.678637\times10^{-33})(5.663217\times10^{93})=1.516970\times10^{61}$$

Y dividimos entre $50!$:

$$
\begin{align*}
P(X=50) &= \frac{1.516970\times10^{61}}{3.041409\times10^{64}}\\
        &= 0.000499
\end{align*}
$$

Por lo tanto:

$$P(X=50)\approx 0.000499$$

Es decir, la probabilidad de que ingresaran exactamente 50 autos en un minuto es de apenas **0.0499%**, mucho menor que la del inciso a). Esto es razonable, ya que 50 está más alejado del valor promedio $\lambda=75$ que 80.

## c) Probabilidad de que en 15 minutos ingresaran 800 autos

Aquí cambia el intervalo de tiempo: ya no es un minuto, sino 15 minutos. Como el parámetro de la distribución de Poisson es proporcional al tamaño del intervalo considerado, hay que reescalar $\lambda$ multiplicándolo por 15:

$$\lambda_{15}=75\times 15=1125\ \text{autos cada 15 minutos}$$

Definiendo:

$$Y=\text{número de autos que ingresan en 15 minutos}, \qquad Y\sim\operatorname{Poisson}(1125)$$

Se busca:

$$P(Y=800)=\frac{e^{-1125}(1125)^{800}}{800!}$$

Calculamos cada parte por separado:

$$e^{-1125}\approx 2.622454\times10^{-489}$$

$$(1125)^{800}\approx 8.356376\times10^{2440}$$

$$800!\approx 7.710530\times10^{1976}$$

Multiplicamos primero el numerador:

$$e^{-1125}(1125)^{800}=(2.622454\times10^{-489})(8.356376\times10^{2440})=2.191421\times10^{1952}$$

Y dividimos entre $800!$:

$$
\begin{align*}
P(Y=800) &= \frac{2.191421\times10^{1952}}{7.710530\times10^{1976}}\\
         &= 2.84\times10^{-25}
\end{align*}
$$

Para dimensionar qué tan pequeño es este número, comparamos 800 con la media y la desviación estándar de $Y$:

$$\mu=E(Y)=\lambda_{15}=1125, \qquad \sigma=\sqrt{\lambda_{15}}=\sqrt{1125}\approx 33.54$$

$$z=\frac{800-1125}{33.54}\approx -9.69$$

Es decir, 800 autos en 15 minutos está casi 10 desviaciones estándar por debajo de lo que se esperaría, lo que explica por qué la probabilidad es prácticamente cero.

Por lo tanto:

$$P(Y=800)\approx 2.84\times10^{-25}$$

Este valor es, en la práctica, cero: bajo el supuesto de que el flujo de autos se mantiene en 75 por minuto, es prácticamente imposible que en 15 minutos ingresaran solo 800 autos, ya que lo esperado serían más de 1100.
