# Problema 1. Salud

De acuerdo con la Encuesta de Consumo de Alcohol, Tabaco y otras Drogas en la Ciudad de México (2009), el 25.3% de las personas consumen marihuana, drogas inhalables o cocaína.

Se tiene un grupo de 25 personas y se desea calcular las probabilidades solicitadas.

## Datos

$$
n=25
$$

$$
p=0.253
$$

$$
q=1-p=0.747
$$

Sea:

$$
X=\text{Número de personas que consumen alguna de estas drogas}
$$

Entonces:

$$
X\sim Binomial(25,0.253)
$$

## a) Probabilidad de que exactamente 10 personas consuman alguna de estas drogas

Se desea calcular:

$$
P(X=10)
$$

Usando la fórmula de la distribución binomial:

$$
P(X=x)=\binom{n}{x}p^x(1-p)^{n-x}
$$

Sustituyendo:

$$
P(X=10)=\binom{25}{10}(0.253)^{10}(0.747)^{15}
$$

```python
from scipy.stats import binom

n = 25
p = 0.253

prob_a = binom.pmf(10, n, p)
print(prob_a)
```

Por lo tanto,

$$
P(X=10)=0.0442
$$

La probabilidad de que exactamente 10 personas consuman alguna de estas drogas es:

$$
4.42\%
$$

## b) Probabilidad de que al menos 20 personas consuman alguna de estas drogas

Se desea calcular:

$$
P(X\geq20)
$$

```python
prob_b = 1 - binom.cdf(19, n, p)
print(prob_b)
```

Por lo tanto,

$$
P(X\geq20)=1.549229167174815\times10^{-8}
$$

La probabilidad es prácticamente cero, es decir, es muy poco probable que al menos 20 de las 25 personas consuman alguna de estas drogas.

## c) Número esperado de personas que consumen alguna de estas drogas entre las siguientes 50 personas

El valor esperado de una distribución binomial se calcula con:

$$
E(X)=np
$$

donde:

$$
n=50
$$

y

$$
p=0.253
$$

```python
esperado = 50 * 0.253
print(esperado)
```

Por lo tanto,

$$
E(X)=50(0.253)=12.65
$$

Se espera que aproximadamente:

$$
13 \text{ personas}
$$

consuman alguna de estas drogas.

## d) Comentario

El consumo de drogas puede afectar la salud física y mental de las personas, además de causar problemas familiares, escolares y sociales.

### Tabla de distribución Binomial

Para complementar el análisis del problema, se construye una tabla de distribución binomial donde se muestran los diferentes valores de X (número de personas consumidoras) y su probabilidad correspondiente.

La distribución binomial se calcula mediante la fórmula:

$$
P(X=x)=\binom{n}{x}p^x(1-p)^{n-x}
$$

Donde:

- $n=25$ representa el número total de personas.
- $p=0.253$ representa la probabilidad de que una persona consuma alguna droga.
- $X$ representa el número de personas que consumen alguna droga.

```python
import pandas as pd
from scipy.stats import binom

# Datos del problema
n = 25
p = 0.253

# Valores posibles de personas consumidoras
x = range(0, n + 1)

# Probabilidades binomiales
probabilidades = binom.pmf(list(x), n, p)

# Crear tabla de distribución
tabla_binomial = pd.DataFrame({
    "Número de personas (X)": x,
    "Probabilidad P(X=x)": probabilidades
})

tabla_binomial
```

### Gráfica de distribución Binomial

La gráfica representa la distribución de probabilidad del número de personas que consumen alguna droga dentro de un grupo de 25 personas. Se observa que los valores cercanos al promedio tienen mayor probabilidad.

```python
import matplotlib.pyplot as plt
from scipy.stats import binom

# Datos del problema
n = 25
p = 0.253

# Valores posibles
x = range(0, n + 1)

# Probabilidades
probabilidades = binom.pmf(list(x), n, p)

# Gráfica
plt.figure(figsize=(10,5))
plt.bar(x, probabilidades)

plt.xlabel("Número de personas consumidoras")
plt.ylabel("Probabilidad")
plt.title("Distribución Binomial (n=25, p=0.253)")

plt.show()
```

# Problema 17. Pasatiempos

Abel Pérez arma rompecabezas como pasatiempo. El rompecabezas que está realizando tiene 5000 piezas y en promedio coloca 100 piezas por día.

Para resolver este problema se utilizará la distribución de Poisson, ya que se quiere conocer la probabilidad de que Abel coloque cierta cantidad de piezas durante un día.

La cantidad promedio de piezas que coloca al día es:

$$
\lambda = 100
$$

## Distribución de Poisson

La fórmula que se utiliza para calcular la probabilidad es:

$$
P(X=x)=\frac{e^{-\lambda}\lambda^x}{x!}
$$

Donde:

$X$ representa la cantidad de piezas que coloca Abel en un día.

$\lambda$ representa el promedio de piezas colocadas por día.

$x$ es la cantidad de piezas que queremos calcular.

```python
from scipy.stats import poisson

media = 100
```

## Inciso a)

Determinar la probabilidad de que Abel coloque entre 100 y 110 piezas en un día, incluyendo ambos valores.

La probabilidad que se desea encontrar es:

$$
P(100 \leq X \leq 110)
$$

Para calcular esta probabilidad se utiliza la función acumulada de Poisson:

$$
P(100 \leq X \leq 110)=P(X\leq110)-P(X\leq99)
$$

```python
# Probabilidad de colocar entre 100 y 110 piezas

prob_a = poisson.cdf(110, media) - poisson.cdf(99, media)

print(prob_a)
```

### Resultado del inciso a)

El resultado obtenido fue:

$$
P(100 \leq X \leq 110)=0.3661614498
$$

Para expresarlo en porcentaje se multiplica por 100:

$$
0.3661614498 \times 100 = 36.62\%
$$

Por lo tanto, la probabilidad de que Abel coloque entre 100 y 110 piezas en un día es aproximadamente:

$$
36.62\%
$$

Esto significa que existe una probabilidad del 36.62% de que Abel coloque esa cantidad de piezas durante un día.

## Inciso b)

Calcular la probabilidad de que Abel coloque exactamente 150 piezas en un día.

La media de piezas colocadas por día es:

$$
\lambda = 100
$$

En este caso buscamos la probabilidad de que la variable tome exactamente el valor de 150:

$$
P(X=150)
$$

Para obtener este valor se utiliza la función de probabilidad de masa de Poisson.

```python
# Probabilidad de colocar exactamente 150 piezas

prob_b = poisson.pmf(150, media)

print(prob_b)
```

### Resultado del inciso b)

El resultado obtenido fue:

$$
P(X=150)=6.511160468786359 \times 10^{-7}
$$

Para convertirlo a porcentaje se multiplica por 100:

$$
(6.511160468786359 \times 10^{-7})(100)
$$

$$
=0.0000651\%
$$

Por lo tanto, la probabilidad de que Abel coloque exactamente 150 piezas en un día es aproximadamente:

$$
0.0000651\%
$$

Esto indica que es un evento poco probable, ya que colocar 150 piezas está por encima del promedio diario de 100 piezas.

## Inciso c)

Calcular la probabilidad de que Abel coloque 45 piezas o menos en un día.

La media de piezas colocadas por día es:

$$

La probabilidad que queremos encontrar es:

### Tabla de distribución de Poisson

Para complementar el análisis del problema, se realiza una tabla de distribución de probabilidad donde se muestran los valores de X (número de piezas colocadas) y su probabilidad correspondiente.

La distribución de Poisson se representa mediante la fórmula:

$$
P(X=x)=\frac{e^{-\lambda}\lambda^x}{x!}
$$

Donde:

- $\lambda = 100$ representa el promedio de piezas colocadas por día.
- $X$ representa el número de piezas colocadas.
- $P(X=x)$ representa la probabilidad de colocar exactamente x piezas.

```python
import pandas as pd
from scipy.stats import poisson

# Datos del problema
lambda_ = 100

# Valores de piezas colocadas
x = range(100, 111)

# Probabilidades
probabilidades = poisson.pmf(list(x), lambda_)

# Crear tabla de distribución
tabla = pd.DataFrame({
    "Número de piezas (X)": x,
    "Probabilidad P(X=x)": probabilidades
})

tabla
```

### Gráfica de distribución de Poisson

La siguiente gráfica representa la distribución de probabilidad del número de piezas colocadas en un día. Se observa que la mayor probabilidad se encuentra cerca de 100 piezas, debido a que este valor representa el promedio diario.

```python
import matplotlib.pyplot as plt
from scipy.stats import poisson

# Datos del problema
lambda_ = 100

# Valores de piezas colocadas
x = range(70, 131)

# Probabilidades de Poisson
probabilidades = poisson.pmf(list(x), lambda_)

# Crear gráfica
plt.figure(figsize=(10,5))
plt.bar(x, probabilidades)

plt.xlabel("Número de piezas colocadas en un día")
plt.ylabel("Probabilidad")
plt.title("Distribución de Poisson (λ = 100)")

plt.show()
```
