```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import binom
n = 15
p = 0.25
x_values = np.arange(0, n + 1)

probabilidades = binom.pmf(x_values, n, p)
```

```python
tabla_distribucion = pd.DataFrame({
    'x (Adolescentes)': x_values,
    'P(X = x)': probabilidades
})

print("--- Distribución de Probabilidad Binomial (n=15, p=0.25) ---")
print(tabla_distribucion.to_string(index=False, formatters={'P(X = x)': '{:.6f}'.format}))
```

```python
plt.figure(figsize=(10, 5))
bars = plt.bar(x_values, probabilidades, color="plum", edgecolor="blue", alpha=0.7)
plt.xticks(x_values)
plt.title(f"Distribución Binomial (n={n}, p={p}) - Cirugía Estética en Adolescentes", fontsize=12, fontweight="bold")
plt.xlabel("Número de adolescentes (x)", fontsize=10)
plt.ylabel("Probabilidad P(X = x)", fontsize=10)
plt.grid(axis='y', linestyle='--', alpha=0.5)

for bar in bars:
    yval = bar.get_height()
    if yval > 0.001:
        plt.text(bar.get_x() + bar.get_width()/2, yval + 0.005, round(yval, 4), ha="center", va="bottom", fontsize=9)

plt.show()

prob_al_menos_10 = 1 - binom.cdf(9, n, p)
```

```python
print("\n==========================================================================")
print("--- Inciso A: Probabilidad de que AL MENOS 10 adolescentes quieran parecerse a un famoso ---")
print(f"P(X >= 10) = 1 - P(X <= 9) = {prob_al_menos_10:.8f}")
print(f"En porcentaje: {prob_al_menos_10 * 100:.6f}%")
print("Nota: Como la probabilidad de éxito (0.25) es baja, la posibilidad de encontrar 10 o más personas de 15 es extremadamente pequeña.")
```

```python
prob_exactamente_5 = binom.pmf(5, n, p)

print("\n==========================================================================")
print("--- Inciso B: Probabilidad de que EXACTAMENTE 5 adolescentes se operen ---")
print(f"P(X = 5) = {prob_exactamente_5:.6f}")
print(f"En porcentaje: {prob_exactamente_5 * 100:.4f}%")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import poisson

lam_3 = 1.0
```

```python
x_values = np.arange(0, 9)
probabilidades = poisson.pmf(x_values, lam_3)
```

```python
tabla_distribucion = pd.DataFrame({
    'x (Autos Robados)': x_values,
    'P(X = x)': probabilidades
})

print("==========================================================================")
print("--- Inciso A: Tabla de Distribución de Probabilidad (Poisson, lambda=1) ---")
print("==========================================================================")
print(tabla_distribucion.to_string(index=False, formatters={'P(X = x)': '{:.6f}'.format}))
```

```python
plt.figure(figsize=(10, 6))
bars = plt.bar(x_values, probabilidades, color='crimson', edgecolor='darkred', alpha=0.7)
plt.xticks(x_values)
plt.title('Distribución de Poisson (λ = 1.0) - Robos de autos en 3 minutos', fontsize=12, fontweight='bold')
plt.xlabel('Número de autos robados (x)', fontsize=10)
plt.ylabel('Probabilidad P(X = x)', fontsize=10)
plt.grid(axis='y', linestyle='--', alpha=0.5)

# Añadimos etiquetas con el valor de probabilidad sobre cada barra
for bar in bars:
    yval = bar.get_height()
    if yval > 0.0001:  # Etiquetar solo valores significativos
        plt.text(bar.get_x() + bar.get_width()/2, yval + 0.005, round(yval, 4), ha='center', va='bottom', fontsize=9)

plt.show()
```

```python
descripcion = """
Descripción de la distribución (Inciso A):
La distribución presenta un marcado SESGO POSITIVO (hacia la derecha). 
Al ser lambda = 1.0 (un promedio bajo), la mayor probabilidad se concentra en valores muy pequeños.
De hecho, hay la misma probabilidad exacta de que no se robe ningún auto (P(X=0) ≈ 36.79%) 
que de que se robe exactamente un auto (P(X=1) ≈ 36.79%). A partir de 2 robos, la probabilidad
disminuye rápidamente debido a la naturaleza de la distribución de Poisson para tasas bajas.
"""
print(descripcion)
```

```python
prob_b = poisson.pmf(2, lam_3)

print("\n==========================================================================")
print("--- Inciso B: Probabilidad de que se roben exactamente 2 autos en 3 min ---")
print("==========================================================================")
print(f"P(X = 2) = {prob_b:.6f}")
print(f"En porcentaje: {prob_b * 100:.4f}%")
```

```python
prob_c = poisson.cdf(2, lam_3)

print("\n==========================================================================")
print("--- Inciso C: Probabilidad de que se roben 2 autos o menos en 3 min ---")
print("==========================================================================")
print(f"P(X <= 2) = {prob_c:.6f}")
print(f"En porcentaje: {prob_c * 100:.4f}%")
```

```python
lam_15 = 5.0
prob_d = poisson.pmf(6, lam_15)

print("\n==========================================================================")
print("--- Inciso D: Probabilidad de que se roben 6 autos en 15 min ---")
print("==========================================================================")
print(f"Para 15 minutos, la nueva tasa media es λ = {lam_15}")
print(f"P(Y = 6) = {prob_d:.6f}")
print(f"En porcentaje: {prob_d * 100:.4f}%")
```
