 <span style="color: blue;">Economía</span>. Una cadena de supermercados, cuyo nombre empieza con W, abrió 267 tiendas en México durante 2010. Con esta información, 
 
 a) Determina λ para el número promedio de tiendas abiertas en un día. 
 
 b) ¿Cuál es la probabilidad de que en un día se abran dos tiendas? 
 
 c) ¿Cuál es la probabilidad de que en una semana se inauguren seis tiendas?

Utilizaremos el modelo de la **Distribución de Poisson**:$$P(X = k) = \frac{e^{-\lambda} \cdot \lambda^k}{k!}$$

**a) Determina $\lambda$ para el número promedio de tiendas abiertas en un día**

Dividimos el total de tiendas inauguradas ($267$) entre los días del año 2010 ($365$ días):$$\lambda_{\text{diario}} = \frac{267}{365} \approx 0.7315 \text{ tiendas por día}$$Respuesta a): El promedio diario de aperturas es $\lambda \approx 0.7315$.

**b) ¿Cuál es la probabilidad de que en un día se abran dos tiendas?**

Utilizamos la fórmula con nuestra tasa diaria $\lambda = 0.7315$ y $k = 2$:$$P(X = 2) = \frac{e^{-0.7315} \cdot (0.7315)^2}{2!} \approx \frac{0.48118 \cdot 0.5351}{2} \approx 0.1287$$Respuesta b): La probabilidad de que en un día se abran exactamente dos tiendas es del $12.87\%$ (o $0.1287$).

**c) ¿Cuál es la probabilidad de que en una semana se inauguren seis tiendas?**

Primero adaptamos el valor de $\lambda$ para un intervalo de una semana (7 días):$$\lambda_{\text{semanal}} = 0.7315 \cdot 7 \approx 5.1205 \text{ tiendas por semana}$$Calculamos la probabilidad para exactamente $k = 6$ aperturas semanales:$$P(X = 6) = \frac{e^{-5.1205} \cdot (5.1205)^6}{6!} \approx \frac{0.005973 \cdot 18,338.31}{720} \approx 0.1521$$Respuesta c): La probabilidad de inaugurar exactamente seis tiendas en una semana es del $15.21\%$ (o $0.1521$).
