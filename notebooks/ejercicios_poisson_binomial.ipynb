## Binomial y Poisson

### Ejercicio 14

<span style="color: blue;">Salud</span>. De acuerdo con la Encuesta de Consumo de Alcohol, Tabaco y otras Drogas en la Ciudad de México (2009), 25.3% de las personas consumen marihuana, drogas inhalables o cocaína. Si tenemos un grupo de 25 personas.

a) ¿Cuál es la probabilidad de que 10 de ellas consuman alguna de estas drogas?

b) ¿Cuál es la probabilidad de que al menos 20 de ellas consuman alguna de estas drogas?

c) De las siguientes 50 personas, ¿cuántas esperarías que consumieran alguna de estas drogas?

Sea:

$$X=\text{número de personas que consumen marihuana, drogas inhalables o cocaína}$$

Como se seleccionan 25 personas y cada una tiene una probabilidad \(p=0.253\) de consumir alguna de estas drogas:

$$X\sim\text{Binomial}(n=25,\;p=0.253)$$

## a) Probabilidad de que exactamente **10 personas** consuman alguna de estas drogas.

$$P(X=10)$$

Usando la fórmula:

$$P(X=x)=\binom{n}{x}p^xq^{n-x}$$

Sustituyendo los datos:

$$P(X=10)=\binom{25}{10}(0.253)^{10}(0.747)^{15}$$
$$\begin{align*}
P(X=10) &= \binom{25}{10}(0.253)^{10}(0.747)^{25-10} \\
        &= \binom{25}{10}(0.253)^{10}(0.747)^{15} \\
        &= 3268760(1.0712\times10^{-6})(0.012626) \\
        &\approx 0.04419
\end{align*}$$

Con ayuda de la tabla de probabilidades obtenemos:

$$P(X=10)\approx0.044197$$

## b) Probabilidad de que al menos 20 personas consuman alguna de estas drogas

Se desea calcular:

$$P(X\ge20)$$

Esto equivale a:

$$\begin{align*}
P(X\ge20)
&=P(X=20)+P(X=21)+P(X=22)+P(X=23)+P(X=24)+P(X=25)
\end{align*}$$

Usando la función de probabilidad binomial:

$$\begin{align*}
P(X\ge20)
&=\sum_{x=20}^{25}\binom{25}{x}(0.253)^x(0.747)^{25-x}\\
&=P(X=20)+P(X=21)+P(X=22)+P(X=23)+P(X=24)+P(X=25)\\
&\approx 0.00000000042
\end{align*}$$

Por lo tanto:

$$P(X\ge20)\approx4.20\times10^{-10}$$

Es decir, la probabilidad es prácticamente igual a:

$$0\%$$

## c) Número esperado de personas que consuman alguna de estas drogas

El valor esperado de una distribución binomial se calcula con:

$$E(X)=np$$

Sustituyendo los datos:

$$\begin{align*}
E(X)
&=(50)(0.253)\\
&=12.65
\end{align*}$$

Por lo tanto:

$$E(X)=12.65$$

Es decir, se espera que aproximadamente:

$$13\text{ personas}$$

consuman alguna de estas drogas.

### Ejercicio 20

<span style="color: blue;">Deportes</span>. En un gimnasio sólo para mujeres se realizan ejercicios aeróbicos y ejercicios en aparatos. El circuito está constituido por nueve aparatos tonificadores y ocho estaciones para los ejercicios aeróbicos. En un día cualquiera, en hora pico (entre las 19 y 20 horas), llegan en promedio siete mujeres.
    

a) ¿Cuál es la probabilidad de que lleguen menos de cinco mujeres?

b) ¿Cuál es la probabilidad de que estén ocupados todos los aparatos?

c) En un día extraordinario, ¿cuál es la probabilidad de que tanto los aparatos como las estaciones estén ocupados?

Sea:

$$X=\text{número de mujeres que llegan al gimnasio entre las 19:00 y las 20:00 horas}$$

Como llegan en promedio 7 mujeres por hora:

$$X\sim\text{Poisson}(\lambda=7)$$

## a) Probabilidad de que lleguen menos de cinco mujeres

Se desea calcular:

$$P(X<5)$$

La expresión "menos de cinco mujeres" significa:

$$P(X<5)$$

Utilizamos la suma de probabilidades:

$$\begin{align*}
P(X<5)
    &=P(X=0)+P(X=1)+P(X=2)+P(X=3)+P(X=4)\\
    &=\frac{e^{-7}(7)^0}{0!}
      +\frac{e^{-7}(7)^1}{1!}
      +\frac{e^{-7}(7)^2}{2!}
      +\frac{e^{-7}(7)^3}{3!}
      +\frac{e^{-7}(7)^4}{4!}\\
    &=0.000912+0.006383+0.022341+0.052129+0.091226\\
    &=0.172992
\end{align*}$$

Por lo tanto:

$$P(X<5)=0.172992$$

La probabilidad de que lleguen menos de cinco mujeres es aproximadamente **17.30%**.

## b) Probabilidad de que estén ocupados todos los aparatos

Como existen nueve aparatos, todos estarán ocupados cuando lleguen exactamente nueve mujeres.

Se desea calcular:

$$P(X=9)$$

Sustituimos $(x=9)$ y $(\lambda=7)$ en la función de probabilidad de Poisson:

$$\begin{align*}
P(X=9)
    &=\frac{e^{-7}(7)^9}{9!}\\
    &=\frac{e^{-7}(40353607)}{362880}\\
    &=0.101405
\end{align*}$$

Por lo tanto:

$$P(X=9)=0.101405$$

La probabilidad de que estén ocupados todos los aparatos es aproximadamente **10.14%**.

## c) Probabilidad de que estén ocupados todos los aparatos y todas las estaciones

Como existen nueve aparatos y ocho estaciones, en total hay:

$$9+8=17$$

lugares disponibles.

La expresión "que tanto los aparatos como las estaciones estén ocupados" significa:

$$P(X\ge17)$$

Utilizamos la regla del complemento:

$$\begin{align*}
P(X\ge17)
    &=1-P(X\le16)\\
    &=1-\left[P(X=0)+P(X=1)+\cdots+P(X=16)\right]\\
    &=1-\left[
        \frac{e^{-7}(7)^0}{0!}
        +\frac{e^{-7}(7)^1}{1!}
        +\cdots
        +\frac{e^{-7}(7)^{16}}{16!}
        \right]\\
    &=1-0.999042\\
    &=0.000958
\end{align*}$$

Por lo tanto:

$$P(X\ge17)=0.000958$$

La probabilidad de que tanto los aparatos como las estaciones estén ocupados es aproximadamente **0.0958%**.
