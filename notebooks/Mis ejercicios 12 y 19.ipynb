<span style="color: blue;">Psicología</span>. Según encuestas realizadas, uno de cada cuatro adolescentes estaría dispuesto a someterse a una cirugía plástica para parecerse a alguna personalidad como David Beckham, Usher o Justin Bieber. Supón que 15 personas asisten a una reunión,

a) ¿Cuál es la probabilidad de que al menos 10 de ellas quisieran parecerse a alguna personalidad?   
b) ¿Cuál es la probabilidad de que exactamente cinco quisieran practicarse una cirugía plástica?   
c) Según estudios psicológicos, esta situación es muy común entre los adolescentes, pues refleja un problema de autoestima y de creer que son valiosos sólo por parecerse a alguien famoso. ¿Consideras que pareciéndose a otra persona uno se pueda sentir contento consigo mismo? ¿Estarías dispuesto a practicarte una cirugía estética para parecerte a tu ídolo? Discute con tus compañeros sobre este tema

Sea

$$
X=\text{Número de adolescentes que desean practicarse una cirugía plástica.}
$$

Como el experimento consiste en observar a $15$ personas y cada una puede presentar únicamente uno de dos posibles resultados:

- Desea practicarse una cirugía plástica.
- No desea practicarse una cirugía plástica.

Además, la probabilidad de éxito permanece constante para todas las personas y se asume que las respuestas son independientes entre sí.

$$
p=\frac{1}{4}=0.25
$$

Por lo tanto, la variable aleatoria $X$ sigue una distribución binomial, es decir,

$$
X\sim\operatorname{Binomial}(n=15,\;p=0.25)
$$

### a) Se desea calcular la probabilidad de que al menos 10 adolescentes deseen practicarse una cirugía plástica.

$$
P(X\ge10).
$$

Los datos para resolver este inciso son

$$
\begin{aligned}
n&=15,\\
p&=\frac{1}{4}=0.25,\\
q&=1-p=\frac{3}{4}=0.75.
\end{aligned}
$$

La función de probabilidad para una distribución binomial está dada por

$$
P(X=x)=\binom{n}{x}p^xq^{\,n-x},
$$

donde:

- $n$ es el número total de ensayos.
- $x$ representa el número de éxitos.
- $p$ es la probabilidad de éxito.
- $q$ es la probabilidad de fracaso.

Se solicita calcular la probabilidad de que al menos 10 adolescentes deseen practicarse una cirugía plástica. Para ello, se calcularán directamente las probabilidades correspondientes a

$$
X=10,\;11,\;12,\;13,\;14,\;15.
$$

Por lo tanto,

$$
P(X\ge10)=P(X=10)+P(X=11)+P(X=12)+P(X=13)+P(X=14)+P(X=15).
$$

### Cálculo de $P(X=10)$

Aplicando la fórmula binomial,

$$
P(X=10)=\binom{15}{10}(0.25)^{10}(0.75)^5.
$$

Primero se calcula la combinación:

$$
\binom{15}{10}
=
\frac{15!}{10!(15-10)!}
=
\frac{15!}{10!5!}
=
3003.
$$

Ahora se calculan las potencias:

$$
(0.25)^{10}\approx0.000001
$$

$$
(0.75)^5\approx0.237305
$$

Sustituyendo los valores obtenidos,

$$
\begin{aligned}
P(X=10)
&=3003(0.000001)(0.237305)\\
&\approx0.0006796.
\end{aligned}
$$

### Cálculo de $P(X=11)$

Aplicando nuevamente la distribución binomial,

$$
P(X=11)=\binom{15}{11}(0.25)^{11}(0.75)^4.
$$

Se calcula la combinación:

$$
\binom{15}{11}
=
\frac{15!}{11!(15-11)!}
=
\frac{15!}{11!4!}
=
1365.
$$

Ahora las potencias:

$$
(0.25)^{11}\approx0.000000
$$

$$
(0.75)^4\approx0.316406
$$

Sustituyendo,

$$
\begin{aligned}
P(X=11)
&=1365(0.000000)(0.316406)\\
&\approx0.0001030.
\end{aligned}
$$

### Cálculo de las probabilidades restantes

Repitiendo el mismo procedimiento para los valores restantes de $X$, se obtiene

$$
P(X=12)=\binom{15}{12}(0.25)^{12}(0.75)^3\approx0.0000114,
$$

$$
P(X=13)=\binom{15}{13}(0.25)^{13}(0.75)^2\approx0.0000009,
$$

$$
P(X=14)=\binom{15}{14}(0.25)^{14}(0.75)\approx0.0000000,
$$

$$
P(X=15)=\binom{15}{15}(0.25)^{15}\approx0.0000000.
$$

Finalmente, se suman todas las probabilidades obtenidas:

$$
\begin{aligned}
P(X\ge10)
&=P(X=10)+P(X=11)+P(X=12)+P(X=13)+P(X=14)+P(X=15)\\
&=0.0006796+0.0001030+0.0000114+0.0000009+0.0000000+0.0000000\\
&\approx0.0007949.
\end{aligned}
$$

Por lo tanto,

$$
\boxed{P(X\ge10)\approx0.0007949.}
$$

Es decir, aproximadamente:
$$8\%$$

### b) ¿Cuál es la probabilidad de que exactamente cinco quisieran practicarse una cirugía plástica?

$$
P(X=5).
$$

Los datos para resolver este inciso son

$$
\begin{aligned}
n&=15,\\
x&=5,\\
p&=\frac{1}{4}=0.25,\\
q&=1-p=\frac{3}{4}=0.75.
\end{aligned}
$$

La función de probabilidad para una distribución binomial está dada por

$$
P(X=x)=\binom{n}{x}p^xq^{\,n-x},
$$

donde:

- $n$ es el número total de ensayos.
- $x$ representa el número de éxitos.
- $p$ es la probabilidad de éxito.
- $q$ es la probabilidad de fracaso.

Se desea calcular la probabilidad de que exactamente cinco adolescentes deseen practicarse una cirugía plástica. Por ello, se sustituye $X=5$ en la función de probabilidad binomial.

Aplicando la fórmula,

$$
P(X=5)=\binom{15}{5}(0.25)^5(0.75)^{15-5}.
$$

Primero se calcula la combinación:

$$
\binom{15}{5}
=
\frac{15!}{5!(15-5)!}
=
\frac{15!}{5!10!}
=
3003.
$$

Ahora se sustituyen los valores obtenidos en la expresión:

$$
\begin{aligned}
P(X=5)
&=3003(0.25)^5(0.75)^{10}\\
&=3003(0.0009765625)(0.0563135147)\\
&\approx0.1651.
\end{aligned}
$$

Por lo tanto,

$$
\boxed{P(X=5)\approx0.1651.}
$$

Es decir, aproximadamente:
$$16.51\%$$

<span style="color: blue;">Economía</span> Una cadena de supermercados, cuyo nombre empieza con W, abrió 267 tiendas en
México durante 2010. Con esta información,

a) Determina $\lambda$ para el número promedio de tiendas abiertas en un día  
b) ¿Cuál es la probabilidad de que en un día se abran dos tiendas?  
c) ¿Cuál es la probabilidad de que en una semana se inauguren seis tiendas?  
d) ¿Qué beneficios económicos trae la inversión de empresas trasnacionales como W? En la actualidad hay una discusión respecto a qué conviene más a las personas: los mercados tradicionales o los supermercados. Junto con tus compañeros, determinen las ventajas y desventajas de cada uno de ellos.

Sea

$$
X=\text{Número de tiendas inauguradas por día.}
$$

Como el problema consiste en analizar el número de tiendas inauguradas durante un intervalo de tiempo determinado, donde los eventos ocurren de manera independiente y se desea conocer el número de ocurrencias en un periodo fijo, la variable aleatoria $X$ sigue una distribución de Poisson.

La función de probabilidad de una distribución de Poisson está dada por

$$
P(X=x)=\frac{e^{-\lambda}\lambda^x}{x!},
$$

donde:

- $X$ representa la variable aleatoria.
- $x$ representa el número de ocurrencias del evento.
- $\lambda$ representa el número promedio de ocurrencias durante el intervalo de tiempo.
- $e$ es la constante de $Euler$, cuyo valor aproximado es $2.71828$.

### a) Determina el valor de $\lambda$ para el número promedio de tiendas abiertas en un día.

De acuerdo con el enunciado, durante el año 2010 se inauguraron 267 tiendas. Como dicho año estuvo conformado por 365 días, el promedio diario de tiendas inauguradas se obtiene mediante la expresión

$$
\lambda=
\frac{\text{Número total de tiendas inauguradas}}
{\text{Número total de días}}.
$$

Sustituyendo los valores del problema,

$$
\lambda=
\frac{267}{365}.
$$

Realizando la división,

$$
\lambda\approx0.7315.
$$

Por lo tanto, la variable aleatoria queda definida como

$$
X\sim\operatorname{Poisson}(0.7315).
$$

### b) ¿Cuál es la probabilidad de que en un día se abran dos tiendas?

$$
P(X=2).
$$

Los datos para resolver este inciso son

$$
\begin{aligned}
x&=2,\\
\lambda&=0.7315.
\end{aligned}
$$

La función de probabilidad de una distribución de Poisson está dada por

$$
P(X=x)=\frac{e^{-\lambda}\lambda^x}{x!},
$$

donde:

- $X$ representa la variable aleatoria.
- $x$ representa el número de ocurrencias del evento.
- $\lambda$ representa el número promedio de ocurrencias durante el intervalo de tiempo.
- $e$ es la constante de Euler, cuyo valor aproximado es $2.71828$.

Como el problema solicita la probabilidad de que se inauguren **exactamente dos tiendas en un día**, se sustituye $x=2$ y $\lambda=0.7315$ en la función de probabilidad de Poisson.

Aplicando la fórmula,

$$
P(X=2)=\frac{e^{-0.7315}(0.7315)^2}{2!}.
$$

Primero se calcula el factorial de 2,

$$
2!=2\times1=2.
$$

Ahora se calcula el cuadrado del valor de $\lambda$,

$$
(0.7315)^2\approx0.5351.
$$

Posteriormente se calcula el valor de la constante de Euler elevada al exponente negativo,

$$
e^{-0.7315}\approx0.4812.
$$

Sustituyendo los valores obtenidos,

$$
P(X=2)=\frac{(0.4812)(0.5351)}{2}.
$$

Se realiza primero la multiplicación del numerador,

$$
(0.4812)(0.5351)\approx0.2575.
$$

Finalmente, se divide entre el factorial calculado,

$$
P(X=2)=\frac{0.2575}{2}\approx0.1287.
$$

Por lo tanto,

$$
\boxed{P(X=2)\approx0.1287.}
$$

Se concluye que la probabilidad de que en un día se inauguren exactamente dos tiendas es aproximadamente $12.87\%$.

### c) ¿Cuál es la probabilidad de que en una semana se inauguren seis tiendas?

$$
P(X=6).
$$

Como el intervalo de tiempo cambia de **un día** a **una semana**, primero es necesario calcular el nuevo valor del parámetro $\lambda$.

Se considera que una semana tiene 7 días, por lo tanto,

$$
\lambda_{\text{semana}}
=
7(0.7315).
$$

Realizando la multiplicación,

$$
\lambda_{\text{semana}}
\approx5.1205.
$$

Los datos para resolver este inciso son

$$
\begin{aligned}
x&=6,\\
\lambda&=5.1205.
\end{aligned}
$$

La función de probabilidad de una distribución de Poisson está dada por

$$
P(X=x)=\frac{e^{-\lambda}\lambda^x}{x!},
$$

donde:

- $X$ representa la variable aleatoria.
- $x$ representa el número de ocurrencias del evento.
- $\lambda$ representa el número promedio de ocurrencias durante el intervalo de tiempo.
- $e$ es la constante de Euler, cuyo valor aproximado es $2.71828$.

Como el problema solicita la probabilidad de que se inauguren **exactamente seis tiendas en una semana**, se sustituyen los valores en la función de probabilidad de Poisson.

Aplicando la fórmula,

$$
P(X=6)
=
\frac{e^{-5.1205}(5.1205)^6}{6!}.
$$

Primero se calcula el factorial,

$$
6!=6\times5\times4\times3\times2\times1=720.
$$

Ahora se calcula la sexta potencia de $\lambda$,

$$
(5.1205)^6\approx18045.4790.
$$

Posteriormente se calcula el valor de la constante de Euler elevada al exponente negativo,

$$
e^{-5.1205}\approx0.0060.
$$

Sustituyendo los valores obtenidos,

$$
P(X=6)
=
\frac{(0.0060)(18045.4790)}{720}.
$$

Se realiza primero la multiplicación del numerador,

$$
(0.0060)(18045.4790)\approx107.6646.
$$

Finalmente, se divide entre el factorial calculado,

$$
P(X=6)
=
\frac{107.6646}{720}
\approx0.1495.
$$

Por lo tanto,

$$
\boxed{P(X=6)\approx0.1495.}
$$

Se concluye que la probabilidad de que en una semana se inauguren exactamente seis tiendas es aproximadamente $14.95\%$.
