sea:  
Probabilidad de exito (p): 0.253  
Probabilidad de fracaso(q=1-p):0.747  
tamaño de la muestra (n): 25 personas  

formula  
$$P(X = k) = \binom{n}{k} \cdot p^k \cdot q^{n-k}$$

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom
from IPython.display import HTML, display

# Parámetros de la distribución binomial actualizados
n = 25
p = 0.253

# Valores de x (de 0 a n)
x = np.arange(0, n + 1)

# Probabilidades P(X = x)
probabilidades = binom.pmf(x, n=n, p=p)

# Construcción de la tabla HTML
html = """
<table style="
    border-collapse: collapse;
    width: 520px;
    text-align: center;
    font-family: Arial, sans-serif;
">
    <thead>
        <tr>
            <th style="
                border: 1px solid #000;
                padding: 10px;
                width: 200px;
            ">x</th>
            <th style="
                border: 1px solid #000;
                padding: 10px;
                width: 320px;
            ">P(X = x)</th>
        </tr>
    </thead>
    <tbody>
"""

for valor, probabilidad in zip(x, probabilidades):
    html += f"""
        <tr>
            <td style="border: 1px solid #000; padding: 8px;">
                {valor}
            </td>
            <td style="border: 1px solid #000; padding: 8px;">
                {probabilidad:.6f}
            </td>
        </tr>
    """

html += """
    </tbody>
</table>
"""

display(HTML(html))
```

## a) ¿Cuál es la probabilidad de que 10 de ellas consuman alguna de estas drogas?

$$P(X = 10) = \binom{25}{10} \cdot (0.253)^{10} \cdot (0.747)^{15}$$

la probabilidad aproximada es de: 3.94%

## b) ¿Cuál es la probabilidad de que al menos 20 de ellas consuman alguna de estas drogas?
$$P(X \ge 20) = P(X=20) + P(X=21) + P(X=22) + P(X=23) + P(X=24) + P(X=25)$$

las probabilidades son muy bajos por que el exito es de 25.3% esta muy lejos de 20 personas de un grupo de 25 por ejemplo x=20

## c) De las siguientes 50 personas, ¿cuántas esperarías que consumieran alguna de estas drogas?

Para calcular el valor esperado usamos la fórmula:$$\mu = n \cdot p$$

entonces:

$$\mu = 50 \cdot 0.253 = 12.65$$

es decir que aproximadamente 13 personas consumen droga
