<h1 align="center">Ejercicio. Distribución Binomial</h1>

**Salud.**  De acuerdo con la Encuesta de Consumo de Alcohol, Tabaco y otras Drogas en la Ciudad de México (2009), 25.3% de las personas consumen marihuana, drogas inhalables o cocaína. Si tenemos un grupo de 25 personas, 


a) ¿Cuál es la probabilidad de que 10 de ellas consuman alguna de estas drogas? 

b) ¿Cuál es la probabilidad de que al menos 20 de ellas consuman alguna de estas drogas? 

c) De las siguientes 50 personas, ¿cuántas esperarías que consumieran alguna de estas drogas?

---

### Identificar el modelo de probabilidad.
Al tratarse de una serie de ensayos independientes (25 personas seleccionadas), donde cada persona solo tiene dos opciones (consumir alguna de estas drogas o no) y cuya probabilidad de éxito es constante ($p = 0.253$), el modelo que mejor describe esta situación es la **distribución binomial**.

La variable aleatoria se define como:
$X =$ número de personas en el grupo que consumen marihuana, drogas inhalables o cocaína.

Los parámetros del modelo son:
* $n = 25$
* $p = 0.253$

Por lo tanto: $X \sim \text{Binomial}(25, 0.253)$

La fórmula general es:
$$P(X = x) = \binom{n}{x} p^x (1-p)^{n-x}$$

Al sustituir los valores del problema:
$$P(X = x) = \binom{25}{x} (0.253)^x (0.747)^{25-x}$$

```python
import matplotlib.pyplot as plt
from scipy.stats import binom
from IPython.display import HTML, display


n = 25
p = 0.253


x = list(range(n + 1))
probabilidades = binom.pmf(x, n, p)



html = """
<table style="border-collapse: collapse; width: 420px; text-align: center; font-family: Arial, sans-serif; border: 1px solid #000;">
    <thead>
        <tr style="background-color: #ffffff;">
            <th style="border: 1px solid #000; padding: 10px; width: 160px; font-weight: bold;">x</th>
            <th style="border: 1px solid #000; padding: 10px; width: 260px; font-weight: bold;">P(X=x)</th>
        </tr>
    </thead>
    <tbody>"""

for k, prob in zip(x, probabilidades):
   
    bg_color = "#f2f2f2" if k % 2 != 0 else "#ffffff"
    html += f"""
        <tr style="background-color: {bg_color};">
            <td style="border: 1px solid #000; padding: 8px;">{k}</td>
            <td style="border: 1px solid #000; padding: 8px;">{prob:.6f}</td>
        </tr>"""

html += """
    </tbody>
</table>
"""


display(HTML(html))
```

```python
import matplotlib.pyplot as plt
from scipy.stats import binom

n = 25
p = 0.253

x = list(range(n + 1))
probabilidades = binom.pmf(x, n, p)


plt.figure(figsize=(8, 5))
plt.bar(x, probabilidades, width=1, edgecolor="w")

plt.title("Distribución binomial: consumo de drogas en CDMX")
plt.xlabel("Número de personas")
plt.ylabel("Probabilidad")
plt.xticks(x)
plt.grid(axis="y", linestyle="--", alpha=0.6)

plt.show()
```

##  ¿Qué tipo de sesgo presenta la distribución?
La distribución presenta un **sesgo positivo (a la derecha)**. Esto ocurre debido a que la probabilidad de éxito ($p = 0.253$) es menor a 0.50, lo que causa que los valores con mayor probabilidad se concentren en el lado izquierdo de la distribución (valores bajos de $X$) y la cola se extienda hacia la derecha (valores altos de $X$).

## a) ¿Cuál es la probabilidad de que 10 de ellas consuman alguna de estas drogas?
Para calcular la probabilidad de que exactamente $X = 10$:

$$P(X = 10) = \binom{25}{10} (0.253)^{10} (0.747)^{25-10}$$

Calculamos el coeficiente binomial:
$$\binom{25}{10} = \frac{25!}{10!(25-10)!} = 3268760$$

Sustituyendo los valores:
$$P(X = 10) = 3268760 \cdot (0.000000099395) \cdot (0.0102657)$$
$$P(X = 10) \approx 0.033351$$

**Respuesta:** La probabilidad de que exactamente 10 de las personas consuman alguna de estas drogas es de **3.34%**.

## b) ¿Cuál es la probabilidad de que al menos 20 de ellas consuman alguna de estas drogas?
La probabilidad de "al menos 20" significa calcular $P(X \geq 20)$. Sumamos las probabilidades desde $X=20$ hasta $X=25$ obtenidas de nuestra tabla de distribución:

$$P(X \geq 20) = P(X = 20) + P(X = 21) + P(X = 22) + P(X = 23) + P(X = 24) + P(X = 25)$$

Extrayendo los valores correspondientes de la tabla de probabilidad:
* $P(X = 20) \approx 0.000000$
* $P(X = 21) \approx 0.000000$
* $P(X = 22) \approx 0.000000$
* $P(X = 23) \approx 0.000000$
* $P(X = 24) \approx 0.000000$
* $P(X = 25) \approx 0.000000$

$$P(X \geq 20) \approx 0.000000$$

**Respuesta:** La probabilidad de que al menos 20 personas consuman alguna de estas drogas es prácticamente **0.00%**.

## c) De las siguientes 50 personas, ¿cuántas esperarías que consumieran alguna de estas drogas?
Para calcular el valor esperado (o media) de una distribución binomial se utiliza la siguiente fórmula, cambiando en este caso el tamaño de la muestra a $n = 50$:

$$E(X) = n \cdot p$$

Sustituyendo nuestros valores:
$$E(X) = 50 \cdot 0.253$$
$$E(X) = 12.65$$

**Respuesta:** De las siguientes 50 personas, se esperaría que **12.65 personas** consumieran alguna de estas drogas.
