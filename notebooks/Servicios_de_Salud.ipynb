De acuerdo con los datos proporcionados por la Secretaría de Salud en 2010, 34% de la población utiliza los servicios de salud públicos federales o estatales. Se quiere realizar una encuesta a 50 personas sobre la calidad de los servicios de salud. Encuentra:

a) ¿Qué proporción de las personas esperas que asistan a los servicios de salud públicos?  
b) ¿Qué probabilidad existe de que 30 personas utilicen los servicios que brinda el gobierno?  
c) ¿Qué probabilidad hay de que cuando mucho 10 utilicen estos servicios?  
d) ¿Alguna vez has recurrido a los servicios públicos de salud? Con tus compañeros de clase comenten sus experiencias y realicen una lluvia de ideas para mejorar los servicios de salud.

Sea:

$$X=\text{número de personas, de las 50 encuestadas, que utilizan los servicios de salud públicos}$$

Se sabe que, de manera histórica, el 34% de la población usa este tipo de servicios. Si suponemos que cada una de las 50 personas encuestadas tiene, de forma independiente, esa misma probabilidad de pertenecer a dicho grupo, entonces $X$ cuenta cuántas de $n$ personas cumplen una condición con probabilidad de éxito fija. Ese es exactamente el escenario de una **distribución binomial**, con parámetros:

$$n=50, \qquad p=0.34$$

Por lo tanto:

$$X\sim\operatorname{Binomial}(n=50,\ p=0.34)$$

La función de probabilidad es:

$$P(X=x)=\binom{50}{x}(0.34)^{x}(0.66)^{50-x}, \qquad x=0,1,2,\ldots,50$$

## a) Proporción esperada de personas que asisten

El 34% que da el enunciado ya es, por sí mismo, la proporción que se esperaría encontrar en cualquier muestra tomada de la población, incluida la de los 50 encuestados. Es decir:

$$\hat{p}=p=0.34=34\%$$

A partir de esta proporción también se puede saber cuántas de las 50 personas encuestadas se espera que efectivamente usen los servicios públicos de salud. Para una distribución binomial, el número esperado de "éxitos" es:

$$E(X)=np$$

Sustituyendo:

$$
\begin{align*}
E(X) &= (50)(0.34)\\
     &= 17
\end{align*}
$$

Entonces se esperaría que, de las 50 personas encuestadas, alrededor de **17** reporten que utilizan los servicios públicos de salud.

## b) Probabilidad de que 30 personas utilicen los servicios del gobierno

Se pide:

$$P(X=30)$$

Sustituimos $x=30$, $n=50$ y $p=0.34$ en la función de probabilidad:

$$P(X=30)=\binom{50}{30}(0.34)^{30}(0.66)^{20}$$

Calculamos cada factor por separado:

$$\binom{50}{30}=4.712921\times10^{13}$$

$$(0.34)^{30}=8.797667\times10^{-15}$$

$$(0.66)^{20}=2.459681\times10^{-4}$$

Multiplicamos primero el coeficiente binomial por $(0.34)^{30}$:

$$\binom{50}{30}(0.34)^{30}=(4.712921\times10^{13})(8.797667\times10^{-15})=0.414627$$

Y este resultado lo multiplicamos por $(0.66)^{20}$:

$$
\begin{align*}
P(X=30) &= (0.414627)(2.459681\times10^{-4})\\
        &= 0.0001020
\end{align*}
$$

Por lo tanto:

$$P(X=30)\approx 0.000102$$

Es decir, la probabilidad de que exactamente 30 de las 50 personas encuestadas usen los servicios públicos de salud es de apenas **0.0102%**, un valor muy bajo. Tiene sentido: en el inciso a) vimos que lo esperado son 17 personas, y 30 está bastante lejos de ese valor.

## c) Probabilidad de que cuando mucho 10 utilicen estos servicios

"Cuando mucho 10" quiere decir que el número de personas que usan el servicio es menor o igual a 10:

$$P(X\le 10)=\sum_{x=0}^{10}\binom{50}{x}(0.34)^{x}(0.66)^{50-x}=\sum_{x=0}^{10}P(X=x)$$

Calculamos cada término de la suma:

$$
\begin{align*}
P(X=0) &= \binom{50}{0}(0.34)^{0}(0.66)^{50} \approx 9.488483\times10^{-10}\\
P(X=1) &= \binom{50}{1}(0.34)^{1}(0.66)^{49} \approx 2.444003\times10^{-8}\\
P(X=2) &= \binom{50}{2}(0.34)^{2}(0.66)^{48} \approx 3.084628\times10^{-7}\\
P(X=3) &= \binom{50}{3}(0.34)^{3}(0.66)^{47} \approx 2.542481\times10^{-6}\\
P(X=4) &= \binom{50}{4}(0.34)^{4}(0.66)^{46} \approx 1.538972\times10^{-5}\\
P(X=5) &= \binom{50}{5}(0.34)^{5}(0.66)^{45} \approx 7.293793\times10^{-5}\\
P(X=6) &= \binom{50}{6}(0.34)^{6}(0.66)^{44} \approx 2.818056\times10^{-4}\\
P(X=7) &= \binom{50}{7}(0.34)^{7}(0.66)^{43} \approx 9.125135\times10^{-4}\\
P(X=8) &= \binom{50}{8}(0.34)^{8}(0.66)^{42} \approx 2.526695\times10^{-3}\\
P(X=9) &= \binom{50}{9}(0.34)^{9}(0.66)^{41} \approx 6.074276\times10^{-3}\\
P(X=10) &= \binom{50}{10}(0.34)^{10}(0.66)^{40} \approx 1.282961\times10^{-2}
\end{align*}
$$

Sumando los once términos:

$$
\begin{align*}
P(X\le 10) &\approx 0.0000000009 + 0.0000000244 + 0.0000003085 + 0.0000025425\\
           &\quad + 0.0000153897+ 0.0000729379 + 0.0002818056 + 0.0009125135\\
           &\quad + 0.0025266950 + 0.0060742760 + 0.0128296100\\
           &= 0.0227161
\end{align*}
$$

Por lo tanto:

$$P(X\le 10)\approx 0.0227$$

Es decir, la probabilidad de que a lo mucho 10 de las 50 personas encuestadas usen los servicios públicos de salud es de aproximadamente **2.27%**. Al igual que en el inciso b), este valor es bajo porque 10 está bastante por debajo del promedio esperado de 17 personas.
