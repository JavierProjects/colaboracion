<span style="color: teal;">Economia</span>. El mar de cortés, en mexico, es la zona donde se localiza el mayor numero de granjas dedicadas al cultivo de ostras perleras. Según los productores, **3.6** de cada cien ostras pueden llegar a tener un buen valor comercial. **Si tenemos 150 ostras** :

a) Determina la probabilidad de que 12 de ellas tengan buen valor comercial.  
b) Encuentra la probabilidad de que cuando mucho dos tengan perlas de buen valor comercial.  

Sea:

$$n=\text{número total de ostras}$$

Ya que tenemos un número fijo de ensayos ($n=150$), dos resultados posibles (tener valor comercial o no) y una probabilidad de éxito constante., utilizamos una **distribución binomial** con una probabilidad de:


$$\text{Prob} = 0.036$$


La función de probabilidad Binomial es:

$$P(X=x)=\binom{n}{x} p^x (1-p)^{n-x}, \quad \mbox{con} \quad x=0,1,2,\ldots,n$$



## a) Probabilidad de que 12 tengan buen valor comercial

Para este problema dado el inciso (a) tenemos :


$$P(X=12)=\binom{150}{12} (0.036)^{12} (0.964)^{150-12}$$

Para resolver esta forfula tenemos que $(150C12 = 1.724206564x10^{17})$ $(p = 4.738381338x10^{-10})$ y $(q = 6.347908799x10^{-3})$. Haciendo la multiplicacion de esos valores tenemos que:



La probabilidad de que exactamente 12 ostras tengan buen valor comercial es aproximadamente del **0.5186%**

## b) Cuando mucho 2

"Cuando mucho 2" significa:
$$P(X \le 2) = P(X=0) + P(X=1) + P(X=2)$$

Para este problema dado el inciso (b) tenemos :


$$P(X=0)=\binom{150}{0} (0.036)^{0} (0.964)^{150-0}$$
$$P(X=0)=4.088416058x10^{-3}$$

$$P(X=1)=\binom{150}{1} (0.036)^{1} (0.964)^{150-1}$$
$$P(X=1)=0.02699$$

$$P(X=2)=\binom{150}{2} (0.036)^{2} (0.964)^{150-2}$$
$$P(X=2)=0.0907$$

Sumando cada una de las probabilidades tenemos que:

La probabilidad de cuando mucho 2 tengan buen valor comercial es **9.07%**

<span style="color: teal;">Pasatiempos</span>. Abel Pérez tiene como pasatiempo armar rompecabezas. Puede pasar horas escuchando música mientras coloca piezas. El rompecabezas que actualmente arma es de 5000 piezas, y en un dia coloca en promedio **100 piezas** :

a) Determina la probabilidad de que en un dia coloque entre 100 y 110 piezas, inclusive.  

b) Calcula la probabilidad de que en un dia en que este muy inspirado coloque 150 piezas.

c) En un dia cualquiera, Abel no está inspirado. ¿Cuál es la probabilidad de que coloque 45 piezas o menos?.


Como menciona un promedio sabemos que:

$$\lambda = 100$$

Por lo tanto:

$$X\sim\operatorname{Poisson}(3.9)$$

La función de probabilidad de Poisson es:

$$P(X=x)=\frac{e^{-\lambda}\lambda^x}{x!}, \quad \mbox{con} \quad x=0,1,2,\ldots$$

## a) Probabilidad de que coloque entre 100 y 110 piezas

Debemos calcular la suma de las probabilidades desde 100 hasta 110:
$$P(100 \le X \le 110) = \sum_{k=100}^{110} \frac{e^{-100} \cdot 100^k}{k!}$$

Aunque la distribución de Poisson es la exacta para este caso, cuando el promedio ($\lambda$) es grande (generalmente mayor a 30), calcular la probabilidad exacta se vuelve muy complicado porque implica sumar muchas factoriales grandes

Utilizando la aproximación normal ya que $\lambda > 30$, donde $\mu = \lambda = 100$ y $\sigma = \sqrt{\lambda} = 10$, aplicamos la corrección de continuidad:

$$P(99.5 \le X \le 110.5) = P\left(\frac{99.5 - 100}{10} \le Z \le \frac{110.5 - 100}{10}\right) = P(-0.05 \le Z \le 1.05)$$

Consultando la tabla Z: $P(Z \le 1.05) - P(Z \le -0.05) \approx 0.8531 - 0.4801 = $ 37.3%.

Por lo tanto la probabilidad de que coloque entre 100 y 110 piezas es: **37.3%**

## b) Probabilidad de 150 piezas

Usando la fórmula de Poisson con $k = 150$:

Tenemos: 

$$P(X = 150) = \frac{e^{-100} \cdot 100^{150}}{150!}$$

La probabilidad de que en un dia este muy inspirado y coloque 150 piezas es **0.0000651%**
