```python
# ==========================
# Gráfica
# ==========================

plt.figure(figsize=(8, 5))
plt.bar(x, probabilidades, width=1, edgecolor="w")

plt.title("Distribución binomial: adolescentes que tuvieron relaciones antes de los 18")
plt.xlabel("Número de adolescentes que tuvieron relaciones antes de los 18")
plt.ylabel("Probabilidad")
plt.xticks(x)
plt.grid(axis="y", linestyle="--", alpha=0.6)
```

## b) Tipo de sesgo

La distribución presenta sesgo a la izquierda, también llamado sesgo negativo.

Esto ocurre porque $p=0.90$, es decir, la probabilidad de éxito es meyor que $0.5$. La mayor parte de la probabilidad se concentra en valores más grandes de $X$, principalmente alrededor de 18 y 19 accidentes, mientras que la cola se extiende hacia valores más pequeños.

## c) Número esperado de adolescentes que hayan tenido relaciones antes de los 18.

Para una distribución binomial:

$$E(X)=np$$

Por lo tanto tenemos:

$$
\begin{align*}
E(X) &= np \\
     &= 20(0.90) \\
     &= 18
\end{align*}
$$

Por lo tanto, se esperarían:

$$18$$

adolescentes hayan tenido relaciones antes de los de la mayoría de edad.

## d) Probabilidad de que 12 adolescentes digan que tuvieron su primera relación sexual antes de los 18 años


$$
\begin{align*}
P(X=5) &= \binom{20}{12}(0.90)^{12}(0.10)^{20-12} \\
       &= \binom{20}{12}(0.90)^{12}(0.10)^8 \\
       &= 125,970(0.282429)(0.00000001) \\
       &= 0.0003557
\end{align*}
$$

Por lo tanto:

$$P(X=12)\approx 0.0003557$$

Es decir, aproximadamente:

$$0.035\%$$

## e) Probabilidad de que más de 17 adolescentes digan que tuvieron relaciones antes de la mayoría de edad 

“Más de 17” significa:

$$P(X >17)$$

Esto quiere decir que buscamos las probabilidades mayores a 17 por eso:

$$P(X > 17) = P(X = 18) + P(X = 19) + P(X = 20)$$

Por lo tanto: 
$$
\begin{align*}
P(X > 17) &=  [P(X=18) + P(X=19)+ P(X=20)] \\
            &= \left[\binom{20}{18}(0.90)^{18}(0.10)^2
            + \binom{20}{19}(0.90)^{19}(0.10)^1 + \binom{20}{20}(0.90)^{20}(0.10)^0\right] \\
            &= \left[(190)(0.15009)(0.10)^2
            + 20(0.13508)(0.10)^1 + (1)(0.12157)(0.10)^0\right] \\
            &= [0.2851 + 0.2701 + 0.1215] \\
            &= 0.6767 \\
\end{align*}
$$

Por lo tanto:

$$P(X > 17) \approx 0.6767$$

Es decir, aproximadamente: 
$$67.67\%$$.

## Ejercicio 17
<span style="color: blue;">Pasatiempos</span>. Abel Pérez tiene como pasatiempo armar rompecabezas. Puede pasar horas escuchando música mientras coloca piezas. El rompecabezas que actualmente arma es de 5000 piezas, y en un día coloca en promedio 100 piezas.

a) Determina la probabilidad de que en un día coloque entre 100 y 110 piezas, inclusive.

b) Calcula la probabilidad de que en un día en que esté muy inspirado coloque 150 piezas.

c) En un día cualquiera, Abel no está inspirado. ¿Cuál es la probabilidad de que coloque 45 o menos?

Sea:

$$X=\text{número de piezas que coloca}$$

Como se conoce el número promedio de piezas que coloca Abel y se desea determinar la probabilidad de que coloque un cierto número de piezas, utilizamos una **distribución de Poisson** con parámetro:


$$\lambda=100$$

Por lo tanto:

$$X\sim\operatorname{Poisson}(10)$$

La función de probabilidad de Poisson es:


$$P(X=x)=\frac{e^{-\lambda}\lambda^x}{x!}, \quad \mbox{con} \quad x=100,101,102,\ldots110$$

Para este problema:


$$P(X=x)=\frac{e^{-100}(100)^x}{x!}$$

## a) Distribución de probabilidad y gráfica

La distribución de Poisson puede tomar una cantidad infinita de valores:


$$x=0,1,2,3,\ldots$$

Para construir una tabla y una gráfica manejables, mostraremos los valores de $(x=80)$ a $(x=110)$. En este intervalo se concentra más del $(90\%)$ de la probabilidad total.

Tabla de distribucion de probabilidad con valores aproximadados:

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson
from IPython.display import HTML, display

# Parámetro de la distribución
lam = 100

# Valores que se mostrarán en la tabla y en la gráfica
x = np.arange(80, 111)

# Probabilidades P(X = x)
probabilidades = poisson.pmf(x, mu=lam)

# Construcción de la tabla HTML
html = """
<table style="
    border-collapse: collapse;
    width: 520px;
    text-align: center;
    font-family: Arial, sans-serif;
">
    <thead>
        <tr>
            <th style="
                border: 1px solid #000;
                padding: 10px;
                width: 200px;
            ">x</th>
            <th style="
                border: 1px solid #000;
                padding: 10px;
                width: 320px;
            ">P(X = x)</th>
        </tr>
    </thead>
    <tbody>
"""

for valor, probabilidad in zip(x, probabilidades):
    html += f"""
        <tr>
            <td style="border: 1px solid #000; padding: 8px;">
                {valor}
            </td>
            <td style="border: 1px solid #000; padding: 8px;">
                {probabilidad:.6f}
            </td>
        </tr>
    """

html += """
    </tbody>
</table>
"""

display(HTML(html))
```

```python
# Gráfica de la distribución
plt.figure(figsize=(9, 5))
plt.bar(x, probabilidades, width=1, edgecolor="w")

plt.title("Distribución de Poisson: piezas colocadas")
plt.xlabel("Número de piezas colocadas")
plt.ylabel("Probabilidad")
plt.xticks(x)
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.tight_layout()

plt.show()
```

## Descripción de la distribución:

La mayor parte de la probabilidad se junta justo en el centro, alrededor de las 99 y 100 piezas.

Esta gráfica se ve casi completamente simétrica (parece una campana parejita) por lo siguiente:

Las barras del lado izquierdo y del lado derecho bajan casi al mismo ritmo a medida que nos alejamos del centro.

No se nota una "cola" más larga para un lado que para el otro; se ve muy bien equilibrada desde la pieza 80 hasta la 110.

Esto pasa porque el promedio del problema es muy alto $(\lambda=100)$. En la distribución de Poisson, cuando el promedio es un número chiquito la gráfica suele quedar muy chueca hacia un lado, pero cuando el promedio es grande como aquí, la distribución se compone y toma la forma de una distribución normal o campana perfecta.

 ## b) Probabilidad de que Abel coloque entre 100 y 110 piezas

Se desea calcular:


$$P(100 \le X \le 110)$$

Sustituimos $(x=100)$ hasta $(x=110)$ y $(\lambda=100)$ en la función de probabilidad de Poisson:

$$
\begin{align*}
P(X=100) &= \frac{e^{-100}(100)^{100}}{100!} = 0.039861  \\
P(X=101) &= \frac{e^{-100}(100)^{101}}{101!} = 0.039466  \\
P(X=102) &= \frac{e^{-100}(100)^{102}}{102!} = 0.038692  \\
P(X=103) &= \frac{e^{-100}(100)^{103}}{103!} = 0.037566  \\
P(X=104) &= \frac{e^{-100}(100)^{104}}{104!} = 0.036121  \\
P(X=105) &= \frac{e^{-100}(100)^{105}}{105!} = 0.034401 \\
P(X=106) &= \frac{e^{-100}(100)^{106}}{106!} = 0.032453  \\
P(X=107) &= \frac{e^{-100}(100)^{107}}{107!} = 0.030330  \\
P(X=108) &= \frac{e^{-100}(100)^{108}}{108!} = 0.028084 \\
P(X=109) &= \frac{e^{-100}(100)^{109}}{109!} = 0.025765  \\
P(X=110) &= \frac{e^{-100}(100)^{110}}{110!} = 0.023423  \\
P(100 \le X \le 110) &= 0.039861 + 0.039466 + 0.038692 + \\
                     &=0.037566 + 0.036121 + 0.034401 + \\
                     &= 0.032453 +  0.030330 + 0.028084 + \\
                     &=0.025765 + 0.023423\\
                     &=0.366163 \\
\end{align*}
$$

Por lo tanto:


$$P(100 \le X \le 110) = 0.366163$$

La probabilidad de Abel coloque entre 100 y 110 piezas es : $36.61\%$.

## c) Probabilidad de Abel este muy inspirado y coloque 150 piezas.

Se desea calcular:


$$P(X=150)$$

Sustituimos $(x=150)$ y $(\lambda=100)$ en la función de probabilidad de Poisson:

$$
\begin{align*}
P(X=150)
    &= \frac{e^{-100}(100)^{150}}{150!} \\
    &= 0.000000651 \\    
\end{align*}
$$

Por lo tanto:


$$P(X=150) = 0.0000651$$

La probabilidad de Abel coloque 150 piezas es de  $0.0000651\%$.

## c) Probabilidad de Abel este muy inspirado y coloque 150 piezas.

Se desea calcular:


$$P(X \leq 45) = \sum_{x=0}^{45} P(X = x)$$

Sustituimos $(x=0)$ hasta $(x=45)$ y $(\lambda=100)$ en la función de probabilidad de Poisson:

$$P(X \leq 45) = \sum_{x=0}^{45} \frac{e^{-100} (100)^x}{x!} = 0.000000$$

Por lo tanto:


$$P(X \leq 45) = 0.000000$$

La probabilidad de Abel coloque 45 piezas o menos es de  $0.0000\%$.
