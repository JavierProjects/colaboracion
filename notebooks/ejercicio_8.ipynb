 <span style="color: blue;">Redes sociales</span>. En México, 6% de los 30 millones de usuarios de internet que hay utilizan Twitter. (Torres, 2010). Si conoces a 15 personas que sean asiduas a las redes sociales: 
 
a) ¿Cuál es la probabilidad de que 9 o más utilicen Twitter? 

b) ¿Cuál es la probabilidad de que máximo dos usen esa red?

**Definimos los parámetros del problema:**

Número de ensayos ($n$): $15$ personas conocidas.

Probabilidad de éxito ($p$): $6\%$, es decir, $p = 0.06$.

Probabilidad de fracaso ($q$): $1 - p = 0.94$.

Fórmula binomial: $P(X = k) = \binom{n}{k} \cdot p^k \cdot q^{n-k}$

**a) ¿Cuál es la probabilidad de que 9 o más utilicen Twitter?**

La frase "9 o más" significa que queremos calcular la probabilidad acumulada desde $k = 9$ hasta $k = 15$:

$$P(X \ge 9) = P(X = 9) + P(X = 10) + P(X = 11) + P(X = 12) + P(X = 13) + P(X = 14) + P(X = 15)$$

Dado que la probabilidad de éxito es muy baja ($0.06$) y el número de personas es $15$, la probabilidad de encontrar a 9 o más usuarios de Twitter en este grupo es extremadamente pequeña. Vamos a calcular los términos más significativos de esta cola de la distribución:

$P(X = 9)$ $= \binom{15}{9} (0.06)^9 (0.94)^6 = 5005 \cdot (1.0078 \times 10^{-11}) \cdot 0.6899 \approx 3.48 \times 10^{-8}$

$P(X = 10)$ $= \binom{15}{10} (0.06)^{10} (0.94)^5 = 3003 \cdot (6.0466 \times 10^{-13}) \cdot 0.7339 \approx 1.33 \times 10^{-9}$

Como puedes observar, ya desde $k = 9$ el valor es de apenas unas cienmillonésimas de unidad, y para valores mayores a 10 la probabilidad es todavía más insignificante (cercana a cero absoluto).

Sumando todos los términos desde $9$ hasta $15$:

$$P(X \ge 9) \approx 3.62 \times 10^{-8} \approx 0.000000036$$

Respuesta: La probabilidad de que 9 o más de las 15 personas utilicen Twitter es prácticamente $0\%$ (específicamente $0.0000036\%$). Esto es completamente lógico, ya que con una tasa del $6\%$, lo esperado en un grupo de 15 personas es que apenas $0.9$ personas (menos de una en promedio) utilicen la plataforma.

**b) ¿Cuál es la probabilidad de que máximo dos usen esa red?**

La expresión "máximo dos" significa que queremos la probabilidad de que $X$ sea menor o igual a 2 ($P(X \le 2)$):$$P(X \le 2) = P(X = 0) + P(X = 1) + P(X = 2)$$Calculamos detalladamente cada uno de los tres escenarios:1. Para $k = 0$ (Ninguno usa Twitter):$$P(X = 0) = \binom{15}{0} \cdot (0.06)^0 \cdot (0.94)^{15}$$$$P(X = 0) = 1 \cdot 1 \cdot 0.3953 = 0.3953$$2. Para $k = 1$ (Exactamente uno usa Twitter):$$P(X = 1) = \binom{15}{1} \cdot (0.06)^1 \cdot (0.94)^{14}$$$$P(X = 1) = 15 \cdot 0.06 \cdot 0.4205 = 0.3785$$3. Para $k = 2$ (Exactamente dos usan Twitter):$$P(X = 2) = \binom{15}{2} \cdot (0.06)^2 \cdot (0.94)^{13}$$$$\binom{15}{2} = \frac{15 \times 14}{2} = 105$$$$P(X = 2) = 105 \cdot 0.0036 \cdot 0.4474 = 0.1691$$4. Sumamos las tres probabilidades:$$P(X \le 2) = 0.3953 + 0.3785 + 0.1691 = 0.9429$$Respuesta b): La probabilidad de que como máximo dos de las quince personas utilicen Twitter es del $94.29\%$ (o $0.9429$). Es un evento sumamente probable debido a la baja tasa de uso general de la plataforma en esa época.
