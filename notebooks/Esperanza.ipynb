```python
# ==========================
# Gráfica
# ==========================

plt.figure(figsize=(8, 5))
plt.bar(x, probabilidades, width=1, edgecolor="w")

plt.title("Distribución binomial: automobiles que cuentan con seguro")
plt.xlabel("Número de autos asegurados")
plt.ylabel("Probabilidad")
plt.xticks(x)
plt.grid(axis="y", linestyle="--", alpha=0.6)

plt.show()
```

## c) ¿Cuál es la probabilidad de que 13 o más autos tengan seguro?

“Al menos dos” significa:

$$P(X \ge 13)$$

Es decir:

$$P(X \ge 13)=[P(X=13)+P(X=4)+P(X=15)+P(X=16)+P(X=17)+P(X=18)]$$

Por lo tanto: 
$$
\begin{align*}
P(X \geq 13) &= [P(X=13)+P(X=4)+P(X=15)+P(X=16)+P(X=17)+P(X=18)]\\
            &= \left[(8568)(0.00005)(0.0418)
            + (3060)(0.00002)(0.0789)
            + (816)(0.00001)(0.1488)
            + (153)(0.000005)(0.2809)
            + (18)(0.000002)(0.53)
            + (1)(0.000001)(1)\right] \\
            &= [0.0343+0.0163+0.0062+0.0017+0.0003+0.0000] \\
            &= 0.0588 \\
\end{align*}
$$

Por lo tanto:

$$P(X \ge 13) \approx  0.0588$$

Es decir, aproximadamente: 
$$5.88\%$$.

<span style="color: blue;">Deportes</span>. En un gimnasio sólo para mujeres se realizan ejercicios aeróbicos y ejercicios en
aparatos. El circuito está constituido por nueve aparatos tonificadores y ocho estaciones para
los ejercicios aeróbicos. En un día cualquiera, en hora pico (entre las 19 y 20 horas), llegan en
promedio siete mujeres. En ese horario:

a) ¿Cuál es la probabilidad de que lleguen menos de cinco mujeres?.

b) ¿Cuál es la probabilidad de que estén ocupados todos los aparatos?.

c) Y en un día extraordinario, ¿cuál es la probabilidad de que tanto los aparatos como las estaciones estén ocupados?.

Como se conoce el número promedio de mujeres que llegan por hora y se desea modelar el número de usuarias que asisten al gimnasio en ese intervalo de tiempo específico, utilizamos una **distribución de Poisson** con parámetro:

$$\lambda=7$$

Por lo tanto:

$$X\sim\operatorname{Poisson}(7)$$

La función de probabilidad de Poisson es:


$$P(X=x)=\frac{e^{-\lambda}\lambda^x}{x!}, \quad \mbox{con} \quad x=0,1,2,3,4\ldots$$

Para este problema:


$$P(X=x)=\frac{e^{-7}(7)^x}{x!}$$

## a) ¿Cuál es la probabilidad de que lleguen menos de cinco mujeres?.

**"Menos de 5"** significa que la cantidad de mujeres que asisten puede ser 0, 1, 2, 3 o 4. no incluye al 5 Para saber la probabilidad total debemos calcular la probabilidad de cada uno de estos escenarios por separado y luego sumarlas.

$$x=0,1,2,3,4$$

$$P(X=0)=\frac{e^{-7}(7)^0}{0!}=\frac{(0.00091188)(1)}{1}=0.00091$$
$$P(X=1)=\frac{e^{-7}(7)^1}{1!}=\frac{(0.00091188)(7)}{1}=0.00638$$
$$P(X=2)=\frac{e^{-7}(7)^2}{2!}=\frac{(0.00091188)(49)}{2}=0.02234$$
$$P(X=3)=\frac{e^{-7}(7)^3}{3!}=\frac{(0.00091188)(343)}{6}=0.05213$$
$$P(X=4)=\frac{e^{-7}(7)^4}{4!}=\frac{(0.00091188)(2401)}{24}=0.09123$$

suma:

$$P(X<5)=0.00091+0.00638+0.02234+0.05213+0.09123$$
$$P(X<5)=0.1730$$


Respuesta: Hay un **17.30%** de probabilidad de que lleguen menos de 5 mujeres.

## b) ¿Cuál es la probabilidad de que estén ocupados todos los aparatos?.

Hay 9 aparatos en total. Para que todos estén ocupados, necesitan llegar 9 o más mujeres
Matemáticamente calcular la probabilidad desde 9 hasta el infinito es imposible Para resolver esto usamos la regla del complemento: sabemos que la probabilidad de todos los escenarios posibles suma 1 (100%) Así que calcularemos lo contrario (la probabilidad de que lleguen entre 0 y 8 mujeres) y se lo restaremos a 1

como:
Ya conocemos la suma del escenario de 0 a 4 mujeres **(0.1730)** solo calculamos con nuestra fórmula los escenarios que faltan: para 5, 6, 7 y 8 mujeres:

$$P(X=10)=\frac{e^{-7}(7)^0}{5!}=0.12772$$
$$P(X=11)=\frac{e^{-7}(7)^1}{6!}=0.14900$$
$$P(X=12)=\frac{e^{-7}(7)^2}{7!}=0.14900$$
$$P(X=13)=\frac{e^{-7}(7)^3}{8!}=0.13038$$

Suma del complemento (de 0 a 8):

 $$(\ge9)=0.1730+0.12772+0.14900+0.14900+0.13038=0.7291$$

Resta final (El complemento):

$$P(X\ge9)=1−0.7291=0.2709$$

Respuesta: Hay una probabilidad del **27.09%** de que se ocupen todos los aparatos.

## c) Y en un día extraordinario, ¿cuál es la probabilidad de que tanto los aparatos como las estaciones estén ocupados?.

Para que todo el gimnasio se sature por completo (tanto aparatos como aeróbicos), se tienen que llenar los 9 aparatos y las 8 estaciones al mismo tiempo. Es decir, se requiere que asistan 17 mujeres o más.

Volvemos a aplicar la regla del complemento: restaremos de 1 la probabilidad de que asistan entre 0 y 16 mujeres 

$$P(\leq16)\approx0.9993$$

Resta final (El complemento):


$$P(\geq17)=1-0.9993=0.0007$$

Respuesta: La probabilidad de un lleno absoluto y extraordinario es de apenas el **0.07%**
