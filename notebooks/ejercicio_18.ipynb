 <span style="color: blue;">Inseguridad</span>. De acuerdo con las estadísticas de criminalidad, en promedio se roba un auto cada tres minutos. Determina: 
 
a) La distribución de probabilidad y la gráfica correspondiente. Describe de manera breve la distribución. 

b) La probabilidad de que en los siguientes tres minutos se roben dos autos. 

c) La probabilidad de que en los siguientes tres minutos se roben dos autos o menos. 

d) La probabilidad de que se roben seis autos en los siguientes 15 minutos.

Sea:

$$X=\text{número de autos robados}$$

Este problema se resuelve con la **distribución de Poisson**, porque se trata de contar el número de robos en un intervalo de tiempo.


$$\lambda=1 robo cada 3 minutos$$

Por lo tanto:

$$X\sim\operatorname{Poisson}(1 robo cada 3 minutos)$$

La función de probabilidad de Poisson es:


$$P(X=x)=\frac{e^{-\lambda}\lambda^x}{x!}, \quad \mbox{con} \quad x=0,1,2,\ldots,5$$

Para este problema:


$$P(X=x)=\frac{e^{-1}(1)^x}{x!}$$

## a) Distribución de probabilidad y gráfica

La distribución de Poisson puede tomar una cantidad infinita de valores:


$$x=0,1,2,3,\ldots,5$$

Para construir una tabla y una gráfica manejables, mostraremos los valores de $(x=0)$ a $(x=5)$.

Tabla de distribucion de probabilidad con valores aproximadados:

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson
from IPython.display import HTML, display

# Parámetro de la distribución
lam = 1

# Valores que se mostrarán en la tabla y en la gráfica
x = np.arange(0, 6)

# Probabilidades P(X = x)
probabilidades = poisson.pmf(x, mu=lam)

# Construcción de la tabla HTML
html = """
<table style="
    border-collapse: collapse;
    width: 520px;
    text-align: center;
    font-family: Arial, sans-serif;
">
    <thead>
        <tr>
            <th style="
                border: 1px solid #000;
                padding: 10px;
                width: 200px;
            ">x</th>
            <th style="
                border: 1px solid #000;
                padding: 10px;
                width: 320px;
            ">P(X = x)</th>
        </tr>
    </thead>
    <tbody>
"""

for valor, probabilidad in zip(x, probabilidades):
    html += f"""
        <tr>
            <td style="border: 1px solid #000; padding: 8px;">
                {valor}
            </td>
            <td style="border: 1px solid #000; padding: 8px;">
                {probabilidad:.6f}
            </td>
        </tr>
    """

html += """
    </tbody>
</table>
"""

display(HTML(html))
```

```python
# Gráfica de la distribución
plt.figure(figsize=(9, 5))
plt.bar(x, probabilidades, width=1, edgecolor="w")

plt.title("Distribución de Poisson: habitantes por vivienda")
plt.xlabel("Número de habitantes por vivienda")
plt.ylabel("Probabilidad")
plt.xticks(x)
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.tight_layout()

plt.show()
```

### Descripción de la distribución

Es una distribución discreta y sesgada a la derecha (con asimetría positiva). Los valores de probabilidad más altos se concentran en $k = 0$ y $k = 1$ (sumando el $73.58\%$ de probabilidad total). A partir de $k = 2$, la probabilidad de que ocurran más robos en ese lapso tan corto de 3 minutos disminuye de forma muy rápida, siendo prácticamente nula para valores mayores a 5.

## b) Probabilidad de que en los siguientes 3 minutos se roben dos autos

Se desea calcular:


$$P(X=2)$$

Sustituimos $(x=5)$ y $(\lambda=3.9)$ en la función de probabilidad de Poisson:

$$
\begin{align*}
P(X=5)
    &= \frac{e^{-1}(1)^2}{2!} \\
    &= \frac{e^{-1}(1)^2}{2} \\
    &= 0.1839 \\    
\end{align*}
$$

Por lo tanto:


$$P(X=5) = 0.1839$$

La probabilidad de que una vivienda sea habitada por cinco personas es aproximadamente $18.39\%$.

## c) La probabilidad de que en los siguientes tres minutos se roben dos autos o menos

La expresión “ los siguientes tres minutos se roben dos autos o menos” significa:

$$P(X \le 2)$$

Utilizamos la regla del complemento:



$$P(X \le 2) = P(X = 0) + P(X = 1) + P(X = 2)$$

$$P(X = 0) = \frac{e^{-1} \cdot 1^0}{0!} \approx 0.3679$$

$$P(X = 1) = \frac{e^{-1} \cdot 1^1}{1!} \approx 0.3679$$

$$P(X = 2) \approx 0.1839$$



Por lo tanto:

**P(X \le 2) \approx 0.3679 + 0.3679 + 0.1839 = 0.9197**

Resultado: La probabilidad es de $91.97\%$ (o $0.9197$).

**d) La probabilidad de que se roben seis autos en los siguientes 15 minutos**

Como el intervalo de tiempo aumentó a 15 minutos (5 veces más que el intervalo base de 3 minutos), nuestra nueva tasa promedio es:$$\lambda = 1 \cdot 5 = 5 \text{ robos cada 15 minutos}$$Calculamos la probabilidad para $k = 6$ usando $\lambda = 5$:$$P(X = 6) = \frac{e^{-5} \cdot 5^6}{6!} = \frac{0.006738 \cdot 15,625}{720} \approx 0.1462$$Resultado: La probabilidad es de $14.62\%$ (o $0.1462$).
