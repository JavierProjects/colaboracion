## a) probabilidad de que lleguen menos de cinco mujeres

La expresión “menos de cinco personas” significa:

$$P(X<5)$$

Utilizamos la regla del complemento:

$$\begin{align*}  P(X < 5) &= P(X \le 4) \\  &= P(X=0) + P(X=1) + P(X=2) + P(X=3) + P(X=4) \\  &= \frac{e^{-7}(7)^0}{0!} + \frac{e^{-7}(7)^1}{1!} + \frac{e^{-7}(7)^2}{2!} + \frac{e^{-7}(7)^3}{3!} + \frac{e^{-7}(7)^4}{4!} \\  &= 0.000912 + 0.006383 + 0.022341 + 0.052129 + 0.091226 \\  &= 0.172992  \end{align*}$$

por lo tanto:

$$P(X<5) = 0.172992$$

la probabilidad de que lleguen menos de cinco mujeres es aproximadamente 
$17.29\%$.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson
from IPython.display import HTML, display

# Parámetro de la distribución
lam = 7

# Valores que se mostrarán en la tabla y en la gráfica
x = np.arange(0, 7)

# Probabilidades P(X = x)
probabilidades = poisson.pmf(x, mu=lam)

# Construcción de la tabla HTML
html = """
<table style="
    border-collapse: collapse;
    width: 520px;
    text-align: center;
    font-family: Arial, sans-serif;
">
    <thead>
        <tr>
            <th style="
                border: 1px solid #000;
                padding: 10px;
                width: 200px;
                background-color: #f2f2f2;
            ">x (Mujeres que llegan)</th>
            <th style="
                border: 1px solid #000;
                padding: 10px;
                width: 320px;
                background-color: #f2f2f2;
            ">P(X = x)</th>
        </tr>
    </thead>
    <tbody>
"""

for valor, probabilidad in zip(x, probabilidades):
    html += f"""
        <tr>
            <td style="border: 1px solid #000; padding: 8px;">
                {valor}
            </td>
            <td style="border: 1px solid #000; padding: 8px;">
                {probabilidad:.6f}
            </td>
        </tr>
    """

html += """
    </tbody>
</table>
"""

display(HTML(html))
```

## b) probabilidad de que estén ocupados todos los aparatos

Asumiendo que las usuarias prefieren ocupar primero los 9 aparatos tonificadores disponibles, requerimos que lleguen exactamente 9 mujeres o mas (X >= 9) = 1 - P(X <= 8). El desglose matemático es:

$$\begin{align*} P(X \le 8) &= P(X=0) + P(X=1) + \dots + P(X=8) \\ &= 0.000912 + 0.006383 + 0.022341 + 0.052129 + 0.091226 + 0.127717 + 0.149003 + 0.149003 + 0.130377 \\ &= 0.729091 \end{align*}$$

por lo tanto:

$$P(X<8) = 27.09$$


## c) En un día extraordinario, ¿cuál es la probabilidad de que tanto los aparatos como las estaciones estén ocupados?

Para que todo el gimnasio esté lleno a su máxima capacidad (9  aparatos + 8 estaciones = 17  lugares), necesitamos que lleguen 17 o más mujeres ($X\ge17$).

$$\begin{align*}  P(X \geq 17) &= 1 - P(X \leq 16) \\  &= 1 - \left[ P(X=0) + P(X=1) + \dots + P(X=16) \right] \\  &= 1 - \left[ \frac{e^{-7}(7)^0}{0!} + \frac{e^{-7}(7)^1}{1!} + \dots + \frac{e^{-7}(7)^{16}}{16!} \right] \\  &= 1 - 0.999042 \\  &= 0.000958\end{align*}$$

Probabilidad de fracaso(q=1-p):0.747

formula  
$$P(X = k) = \binom{n}{k} \cdot p^k \cdot q^{n-k}$$

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom
from IPython.display import HTML, display

# Parámetros de la distribución binomial actualizados
n = 25
p = 0.253

# Valores de x (de 0 a n)
x = np.arange(0, n + 1)

# Probabilidades P(X = x)
probabilidades = binom.pmf(x, n=n, p=p)

# Construcción de la tabla HTML
html = """
<table style="
    border-collapse: collapse;
    width: 520px;
    text-align: center;
    font-family: Arial, sans-serif;
">
    <thead>
        <tr>
            <th style="
                border: 1px solid #000;
                padding: 10px;
                width: 200px;
            ">x</th>
            <th style="
                border: 1px solid #000;
                padding: 10px;
                width: 320px;
            ">P(X = x)</th>
        </tr>
    </thead>
    <tbody>
"""

for valor, probabilidad in zip(x, probabilidades):
    html += f"""
        <tr>
            <td style="border: 1px solid #000; padding: 8px;">
                {valor}
            </td>
            <td style="border: 1px solid #000; padding: 8px;">
                {probabilidad:.6f}
            </td>
        </tr>
    """

html += """
    </tbody>
</table>
"""

display(HTML(html))
```
