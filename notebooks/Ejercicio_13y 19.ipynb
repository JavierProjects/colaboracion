# Ejercicio 13
El Mar de Cortés, en México, es la zona donde se localiza el mayor número de granjas dedicadas al cultivo de ostras perleras. Según los productores, **3.6 de cada cien ostras** pueden llegar a tener un buen valor comercial. Si tenemos **150 ostras**:

a) Determina la probabilidad de que 12 de ellas tengan perlas de buen valor comercial.  
b) Encuentra la probabilidad de que cuando mucho dos tengan perlas de buen valor comercial.

Sea:

$$X=\text{número de ostras que tienen perlas de buen valor comercial}$$

Se tiene un número fijo de ostras:

$$n=150$$

La probabilidad de que una ostra tenga una perla de buen valor comercial es:

$$p=\frac{3.6}{100}=0.036$$

La probabilidad de que no tenga una perla de buen valor comercial es:

$$q=1-p=1-0.036=0.964$$

Por lo tanto:

$$X\sim\operatorname{Binomial}(n=150,p=0.036)$$

La función de probabilidad binomial es:

$$P(X=x)=\binom{n}{x}p^xq^{n-x}$$

Para este problema:

$$P(X=x)=\binom{150}{x}(0.036)^x(0.964)^{150-x}$$

## a) Probabilidad de que exactamente 12 tengan perlas de buen valor comercial

Se desea calcular:

$$P(X=12)$$

Sustituimos los valores:

$$
\begin{align*}
P(X=12)
    &= \binom{150}{12}(0.036)^{12}(0.964)^{150-12} \\
    &= \binom{150}{12}(0.036)^{12}(0.964)^{138} \\
    &\approx 0.005186
\end{align*}
$$

En porcentaje:

$$
\boxed{P(X=12)\approx 0.5186\%}
$$

## b) Probabilidad de que cuando mucho dos tengan perlas de buen valor comercial

“Cuando mucho dos” significa:

$$X\leq 2$$

Por lo tanto:

$$P(X\leq 2)=P(X=0)+P(X=1)+P(X=2)$$

Sustituimos:

$$
\begin{align*}
P(X\leq 2)
    &= \binom{150}{0}(0.036)^0(0.964)^{150} \\
    &\quad +\binom{150}{1}(0.036)^1(0.964)^{149} \\
    &\quad +\binom{150}{2}(0.036)^2(0.964)^{148} \\
    &\approx 0.004088+0.022902+0.063717 \\
    &\approx 0.090707
\end{align*}
$$

En porcentaje:

$$
\boxed{P(X\leq 2)\approx 9.0707\%}
$$

## Resultados

$$
\boxed{\text{a) }P(X=12)\approx 0.005186}
$$

$$
\boxed{\text{b) }P(X\leq 2)\approx 0.090707}
$$

# Ejercicio 19

Una cadena de supermercados abrió **267 tiendas en México durante 2010**.

**a)** Determina $\lambda$ para el número promedio de tiendas abiertas en un día.  

**b)** ¿Cuál es la probabilidad de que en un día se abran dos tiendas?  

**c)** ¿Cuál es la probabilidad de que en una semana se inauguren seis tiendas?

## Datos

Durante 2010 se abrieron:

$$
267\text{ tiendas}
$$

Como el año 2010 tuvo 365 días:

$$
365\text{ días}
$$

Para resolver el ejercicio se utiliza la distribución de Poisson:

$$
P(X=x)=\frac{e^{-\lambda}\lambda^x}{x!}
$$

## a) Valor de $\lambda$ por día

El promedio de tiendas abiertas en un día es:

$$
\lambda=\frac{267}{365}
$$

$$
\lambda\approx 0.7315
$$

Por lo tanto:

$$
\boxed{\lambda\approx 0.7315\text{ tiendas por día}}
$$

## b) Probabilidad de que en un día se abran dos tiendas

Se desea calcular:

$$
P(X=2)
$$

Sustituyendo en la fórmula de Poisson:

$$
P(X=2)=\frac{e^{-0.7315}(0.7315)^2}{2!}
$$

$$
P(X=2)\approx 0.128741
$$

En porcentaje:

$$
\boxed{P(X=2)\approx 12.87\%}
$$

## c) Probabilidad de que en una semana se inauguren seis tiendas

Primero se calcula el promedio para una semana:

$$
\lambda_{\text{semana}}=7(0.7315)
$$

$$
\lambda_{\text{semana}}\approx 5.1205
$$

Ahora se desea calcular:

$$
P(X=6)
$$

Sustituyendo en la fórmula de Poisson:

$$
P(X=6)=\frac{e^{-5.1205}(5.1205)^6}{6!}
$$

$$
P(X=6)\approx 0.149534
$$

En porcentaje:

$$
\boxed{P(X=6)\approx 14.95\%}
$$

## Resultados

$$
\boxed{\text{a) }\lambda\approx 0.7315}
$$

$$
\boxed{\text{b) }P(X=2)\approx 0.128741}
$$

$$
\boxed{\text{c) }P(X=6)\approx 0.149534}
$$
