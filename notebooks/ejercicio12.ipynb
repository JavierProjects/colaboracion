Psicología. Según encuestas realizadas, uno de cada cuatro adolescentes estaría dispuesto a someterse a una cirugía plástica para parecerse a alguna personalidad como David Beckham, Usher o Justin Bieber. Supón que 15 personas asisten a una reunión

a) ¿Cuál es la probabilidad de que al menos 10 de ellas quisieran parecerse a alguna personalidad?

b) ¿Cuál es la probabilidad de que exactamente cinco quisieran practicarse una cirugía plástica?

Sea:

$$X=\mbox{El número de adolescentes que están dispuestos a someterse a una cirugía plástica para parecerse a una celebridad}$$

sabiendo que la probabilidad de éxito es de 1 de cada 4 = 0.25

$$X∼\mbox{Binomial}(n=15,p=0.25)$$

la funcion que utilizaremos es:
$$P(X=x)=\binom{15}{x}(0.25)^x(0.75)^{15-x} \quad \mbox{con} \quad x=0,1,2,\dots,10$$

## a) Probabilidad de que al menos 10 adolescentes la deseen

Para calcular la probabilidad de que X sea mayor o igual a 10, es decir, $P(X \ge 10)$, utilizamos la regla del complemento restando de 1 la suma de todas las probabilidades individuales menores a 10 (desde x = 0 hasta x = 9):

Es decir:
$P(X \ge 10) = 1 - P(X < 10)$

Por lo tanto:
$P(X \ge 10) = 1 - [P(X = 0) + P(X = 1) + P(X = 2) + \dots + P(X = 9)]$

Sustituyendo cada probabilidad individual calculada con la fórmula binomial $\binom{15}{x}(0.25)^x(0.75)^{15-x}$:

* $P(X = 0) = \binom{15}{0}(0.25)^0(0.75)^{15} = 0.013363$
* $P(X = 1) = \binom{15}{1}(0.25)^1(0.75)^{14} = 0.066817$
* $P(X = 2) = \binom{15}{2}(0.25)^2(0.75)^{13} = 0.155907$
* $P(X = 3) = \binom{15}{3}(0.25)^3(0.75)^{12} = 0.225199$
* $P(X = 4) = \binom{15}{4}(0.25)^4(0.75)^{11} = 0.225199$
* $P(X = 5) = \binom{15}{5}(0.25)^5(0.75)^{10} = 0.165146$
* $P(X = 6) = \binom{15}{6}(0.25)^6(0.75)^9 = 0.091748$
* $P(X = 7) = \binom{15}{7}(0.25)^7(0.75)^8 = 0.039321$
* $P(X = 8) = \binom{15}{8}(0.25)^8(0.75)^7 = 0.013107$
* $P(X = 9) = \binom{15}{9}(0.25)^9(0.75)^6 = 0.003398$

Agrupando los términos para obtener la suma de la probabilidad acumulada:
$P(X \ge 10) = 1 - [0.013363 + 0.066817 + 0.155907 + 0.225199 + 0.225199 + 0.165146 + 0.091748 + 0.039321 + 0.013107 + 0.003398]$

$P(X \ge 10) = 1 - 0.999206$

Por lo tanto:

$$P(X \ge 10) \approx 0.000794$$

Es decir, aproximadamente:

$$0.0794\%$$

## b) Probabilidad de que exactamente cinco la deseen

$$P(X = 5) = \binom{15}{5} (0.25)^5 (0.75)^{15-5}$$
$$= \binom{15}{5} (0.25)^5 (0.75)^{10}$$
$$= 3003 \cdot (0.00097656) \cdot (0.05631351)$$
$$= 0.1651$$

Por lo tanto:

$$P(X = 5) \approx 0.1651$$

Es decir, aproximadamente:

$$16.51\%$$
