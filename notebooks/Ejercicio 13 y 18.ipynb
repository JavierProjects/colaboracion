### Identificación de la distribución

Este experimento cumple las características de una **distribución binomial**, ya que:

- Se analiza un número fijo de ostras ($n=150$).
- Cada ostra solo puede producir dos resultados: tener una perla de valor comercial o no tenerla.
- La probabilidad de éxito permanece constante para todas las ostras.
- Se considera que el resultado obtenido por una ostra no modifica el de las demás.

Los parámetros del problema son:

$$
n=150
$$

$$
p=0.036
$$

$$
q=1-p=0.964
$$

Por lo tanto,

$$
X \sim B(150,\,0.036)
$$

### a) Probabilidad de que exactamente 12 ostras tengan perlas de buen valor comercial

Datos:

$$
n=150,\qquad x=12,\qquad p=0.036,\qquad q=0.964
$$

Se aplica la distribución binomial:

$$
P(X=x)=\binom{n}{x}p^xq^{\,n-x}
$$

Sustituyendo los datos:

$$
P(X=12)=\binom{150}{12}(0.036)^{12}(0.964)^{150-12}
$$

Se resuelve el exponente:

$$
150-12=138
$$

Entonces,

$$
P(X=12)=\binom{150}{12}(0.036)^{12}(0.964)^{138}
$$

Evaluando la expresión con Python:

$$
P(X=12)=0.0051862086
$$

Por lo tanto,

$$
P(X=12)\approx0.00519
$$

Es decir, aproximadamente:

$$
0.52\%
$$

### b) Probabilidad de que cuando mucho dos ostras tengan perlas de buen valor comercial

La expresión **cuando mucho dos** significa:

$$
P(X\le2)=P(X=0)+P(X=1)+P(X=2)
$$

Aplicando la distribución binomial para cada caso:

$$
P(X=0)=\binom{150}{0}(0.036)^0(0.964)^{150}
$$

$$
P(X=0)\approx0.004083
$$

$$
P(X=1)=\binom{150}{1}(0.036)^1(0.964)^{149}
$$

$$
P(X=1)\approx0.022868
$$

$$
P(X=2)=\binom{150}{2}(0.036)^2(0.964)^{148}
$$

$$
P(X=2)\approx0.063756
$$

Se suman las probabilidades:

$$
P(X\le2)=0.004083+0.022868+0.063756
$$

$$
P(X\le2)=0.090707
$$

Por lo tanto,

$$
P(X\le2)\approx0.09071
$$

Es decir, aproximadamente:

$$
9.07\%
$$

### Distribución de Poisson

El número de robos de automóviles se modela mediante una distribución de Poisson, ya que se estudia el número de eventos que ocurren durante un intervalo de tiempo conocido.

La función de probabilidad es

$$
P(X=x)=\frac{e^{-\lambda}\lambda^x}{x!}
$$

### a) Distribución de probabilidad

Datos:

$$
\lambda=1
$$

Se calcula la distribución de Poisson para los robos ocurridos en un intervalo de tres minutos.

La siguiente tabla muestra la distribución de probabilidad.

```python
from scipy.stats import poisson
from IPython.display import HTML
import matplotlib.pyplot as plt
import numpy as np

lam = 1
```

```python
from IPython.display import HTML

x = list(range(8))
prob = poisson.pmf(x, lam)

html="""
<table style="border-collapse:collapse;text-align:center;">
<tr>
<th style="border:1px solid black;padding:8px;">x</th>
<th style="border:1px solid black;padding:8px;">P(X=x)</th>
</tr>
"""

for i,p in zip(x,prob):
    html+=f"""
<tr>
<td style="border:1px solid black;padding:8px;">{i}</td>
<td style="border:1px solid black;padding:8px;">{p:.6f}</td>
</tr>
"""

html+="</table>"

HTML(html)
```

```python
plt.figure(figsize=(7,4))
plt.bar(x,prob)
plt.title("Distribución de Poisson")
plt.xlabel("Número de robos")
plt.ylabel("Probabilidad")
plt.grid(axis="y")
plt.show()
```

La distribución presenta su mayor probabilidad en un robo durante tres minutos y las probabilidades disminuyen conforme aumenta el número de robos.

### b) Probabilidad de que en los siguientes tres minutos se roben dos autos
