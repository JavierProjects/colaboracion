<span style="color: blue;">Sexualidad</span>. Con base en la Encuesta Nacional de Juventud 2005, 90% de los adolescentes entre 15 y 19 años afirmó que tuvo su primera relación sexual antes de los 18 años. Si en un grupo hay 20 adolescentes entre 15 y 19 años, entonces:

a) ¿Cuántos esperarías que contestaran que tuvieron su primera relación sexual antes de       los 18 años?

b) ¿Cuál es la probabilidad de que 12 de ellos digan que tuvieron su primera relación         sexual antes de los 18 años?

c) ¿Cuál es la probabilidad de que más de 17 digan que tuvieron su primera relación           sexual antes de los 18 años?

Sea:

$$X=\mbox{Número de adolescentes que tuvieron relaciones antes de los 18 años}$$

Hay un total de 20 alumnos con una probabilidad p=0.90 de decir Sí he tenido relaciones antes de los 18:

$$X∼\mbox{Binomial}(n=20,p=0.90)$$





## a) Distribución de probabilidad

La función es:

$$P(X=x)=\binom{20}{x}(0.90)^x(0.10)^{20-x} \quad \mbox{con} \quad x=0,1,2,\dots,20$$

```python
import matplotlib.pyplot as plt
from scipy.stats import binom

# Parámetros
n = 20
p = 0.90

# Valores posibles de X
x = list(range(n + 1))

# Distribución binomial
probabilidades = binom.pmf(x, n, p)

# Mostrar tabla en consola
#for k, prob in zip(x, probabilidades):
#    print(f"P(X = {k}) = {prob:.6f}")

# ==========================
# Generar tabla HTML
# ==========================

html = """
<table style="border-collapse: collapse; width: 420px; text-align: center;">
  <thead>
    <tr>
      <th style="border: 1px solid #000; padding: 10px; width: 160px;">x</th>
      <th style="border: 1px solid #000; padding: 10px; width: 260px;">P(X=x)</th>
    </tr>
  </thead>
  <tbody>"""

for k, prob in zip(x, probabilidades):
    html += f"""
    <tr>
      <td style="border: 1px solid #000; padding: 8px;">{k}</td>
      <td style="border: 1px solid #000; padding: 8px;">{prob:.20f}</td>
    </tr>"""

html += """
  </tbody>
</table>
"""

# Mostrar el HTML en la consola
print("\nCódigo HTML:\n")
print(html)

```

```python
# ==========================
# Gráfica
# ==========================

plt.figure(figsize=(8, 5))
plt.bar(x, probabilidades, width=1, edgecolor="w")

plt.title("Distribución binomial: adolescentes que tuvieron relaciones antes de los 18")
plt.xlabel("Número de adolescentes que tuvieron relaciones antes de los 18")
plt.ylabel("Probabilidad")
plt.xticks(x)
plt.grid(axis="y", linestyle="--", alpha=0.6)
```

## b) Tipo de sesgo

La distribución presenta sesgo a la izquierda, también llamado sesgo negativo.

Esto ocurre porque $p=0.90$, es decir, la probabilidad de éxito es meyor que $0.5$. La mayor parte de la probabilidad se concentra en valores más grandes de $X$, principalmente alrededor de 18 y 19 accidentes, mientras que la cola se extiende hacia valores más pequeños.

## c) Número esperado de adolescentes que hayan tenido relaciones antes de los 18.

Para una distribución binomial:

$$E(X)=np$$

Por lo tanto tenemos:

$$
\begin{align*}
E(X) &= np \\
     &= 20(0.90) \\
     &= 18
\end{align*}
$$

Por lo tanto, se esperarían:

$$18$$

adolescentes hayan tenido relaciones antes de los de la mayoría de edad.

## d) Probabilidad de que 12 adolescentes digan que tuvieron su primera relación sexual antes de los 18 años


$$
\begin{align*}
P(X=5) &= \binom{20}{12}(0.90)^{12}(0.10)^{20-12} \\
       &= \binom{20}{12}(0.90)^{12}(0.10)^8 \\
       &= 125,970(0.282429)(0.00000001) \\
       &= 0.0003557
\end{align*}
$$

Por lo tanto:

$$P(X=12)\approx 0.0003557$$

Es decir, aproximadamente:

$$0.035\%$$

## e) Probabilidad de que más de 17 adolescentes digan que tuvieron relaciones antes de la mayoría de edad 

“Más de 17” significa:

$$P(X >17)$$

Esto quiere decir que buscamos las probabilidades mayores a 17 por eso:

$$P(X > 17) = P(X = 18) + P(X = 19) + P(X = 20)$$

Por lo tanto: 
$$
\begin{align*}
P(X > 17) &=  [P(X=18) + P(X=19)+ P(X=20)] \\
            &= \left[\binom{20}{18}(0.90)^{18}(0.10)^2
            + \binom{20}{19}(0.90)^{19}(0.10)^1 + \binom{20}{20}(0.90)^{20}(0.10)^0\right] \\
            &= \left[(190)(0.15009)(0.10)^2
            + 20(0.13508)(0.10)^1 + (1)(0.12157)(0.10)^0\right] \\
            &= [0.2851 + 0.2701 + 0.1215] \\
            &= 0.6767 \\
\end{align*}
$$

Por lo tanto:

$$P(X > 17) \approx 0.6767$$

Es decir, aproximadamente: 
$$67.67\%$$.
