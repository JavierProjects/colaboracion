<span style="color: brown;">Redes Sociales</span>. En México, 6% de los 30 millones de usuarios de internet que hay utilizan Twitter.
Si conoces a 15 personas que sean asiduas a las redes sociales:  

a) ¿Cuál es la probabilidad de que 9 o más utilicen Twitter?

b) ¿Cuál es la probabilidad de que máximo dos usen esa red?

Sea:

$$X=\mbox{Número de usuarios de internet (de entre las 15 personas conocidas) que asiduamente utilizan Twitter.}$$

Como hay 15 personas y cada una tiene una probabilidad p = 0.06 de ser un usuario de Twitter:

$$X∼\mbox{Binomial}\quad(n=15,p=0.06)$$

## a) Probabilidad de que 9 o más utilicen Twitter

Para encontrar la probabilidad de que 9 o más personas utilicen Twitter, necesitamos calcular:
$$P(X \geq 9)=P(X=9)+P(X=10)+P(X=11)+P(X=12)+P(X=13)+P(X=14)+P(X=15)$$

Dado que la probabilidad de éxito es muy baja (6%) y la muestra es pequeña (15), la probabilidad de encontrar a 9 o más personas es extremadamente cercana a cero. 
Por ello, calculamos el primer término ($X=9$):

$$
\begin{align*}
P(X=9) &= \binom{15}{9}(0.06)^9(0.94)^{15-9} \\
       &= \binom{15}{9}(0.06)^9(0.94)^6 \\
       &= 5005(1.0077696*10^{-11})(0.689869) \\
       &= 3.4796 * 10^{-8}
\end{align*}
$$

Por lo tanto:

$$P(X=9)\approx 3.4796 * 10^{-8}$$

Es decir, aproximadamente:

$$0.0000034796\%$$

Al sumar el resto de los términos hasta 15, el resultado sigue siendo insignificante: 

Probabilidad: $\approx 0.000000035$ (o lo que es lo mismo, 0.0000035%).

## b)  Probabilidad de que máximo dos usen esa red

“Máximo dos” significa:

$$P(X \le 2)$$

Lo que quiere decir que la red sea usada por 0, 1 o 2 personas; por lo tanto, usaremos la siguiente fórmula:

$$P(X \le 2) = P(X = 0)+P(X = 1)+P(X = 2) $$

Por lo tanto, calculamos cada término por separado: 
$$
\begin{align*}
P(X = 0)&= \binom{n}{x}(p)^x(q)^{n-x}\\
        &= \binom{15}{0}(0.06)^0(0.94)^{15-0} \\
        &= \binom{15}{0}(0.06)^0(0.94)^{15} \\
        &= 1(1)(0.395292) \\
        &= 0.3953
\end{align*}
$$

$$
\begin{align*}
P(X = 1)&= \binom{n}{x}(p)^x(q)^{n-x}\\
        &= \binom{15}{1}(0.06)^1(0.94)^{15-1} \\
        &= \binom{15}{1}(0.06)^1(0.94)^{14} \\
        &= 15(0.06)(0.420523) \\
        &= 0.3785
\end{align*}
$$

$$
\begin{align*}
P(X = 2)&= \binom{n}{x}(p)^x(q)^{n-x}\\
        &= \binom{15}{2}(0.06)^2(0.94)^{15-2} \\
        &= \binom{15}{2}(0.06)^2(0.94)^{13} \\
        &= 105(0.0036)(0.447365) \\
        &= 0.1691
\end{align*}
$$

Lo sumamos:
$$
\begin{align*}
P(X \le 2)&= P(X = 0)+P(X = 1)+P(X = 2)\\
          &= (0.3953)+(0.3785)+(0.1691)\\
          &= 0.9429
\end{align*}
$$

Por lo tanto:

$$P(X \le 2) \approx 0.9429$$

Es decir, aproximadamente: 
$$94.29\%$$.

<span style="color: brown;">Turismo</span>. Al término de la Semana Santa de 2011, en la caseta de la autopista a Cuernavaca
ingresaban 75 autos en promedio por minuto. Con este dato, calcula la probabilidad de que en
un minuto dado, ingresaran:  
   
a) 80 autos   
b) 50 autos   
c) De que en 15 minutos, ingresaran 800 autos

Sea:

$$X=\mbox{Número de automóviles que pasan por la caseta de la autopista en el intervalo de un minuto.}$$

Como se conoce el número promedio de autos por minuto y se desea modelar el número de vehículos que ingresan en ese intervalo de tiempo, utilizamos una **distribución de Poisson** con parámetro:

$$\lambda=75$$

Por lo tanto:

$$X\sim\operatorname{Poisson}(75)$$

La función de probabilidad de Poisson es:


$$P(X=x)=\frac{e^{-\lambda}\lambda^x}{x!}, \quad \mbox{con} \quad x=0,1,2,\ldots$$

Donde $e \approx 2.71828$ es la base de los logaritmos naturales (constante de Euler).


Para este problema:


$$P(X=x)=\frac{e^{-75}(75)^x}{x!}$$

## a)  Probabilidad de que en un minuto dado, ingresaran 80 autos

Se desea calcular:


$$P(X=80)$$

Sustituimos $(x=80)$ y $(\lambda=75)$ en la función de probabilidad de Poisson:

$$
\begin{align*}
P(X=80)
    &= \frac{e^{-75}(75)^{80}}{80!} \\
    &= \frac{2.678637 \times 10^{-33}(1.011342)^{150}}{7.156946*10^{118}} \\
    &= 0.037852
 \\    
\end{align*}
$$

Por lo tanto:


$$P(X=80) = 0.037852$$

La probabilidad de que en un minuto dado, ingresen 80 autos es de aproximadamente $3.8\%$.

## b) Probabilidad de que en un minuto dado, ingresaran 50 autos

Se desea calcular:


$$P(X=50)$$

Sustituimos $(x=50)$ y $(\lambda=75)$ en la función de probabilidad de Poisson:

$$\begin{align*}
P(X=50) &= \frac{e^{-75}(75)^{50}}{50!} \\
&= \frac{2.678637 \times 10^{-33} (5.663185) \times 10^{93}}{3.041409 \times 10^{64}} \\
&= \frac{1.5169 \times 10^{61}}{3.0414 \times 10^{64}} \\
&= 0.000499
\end{align*}$$

Por lo tanto:


$$P(X=50) = 0.000499$$

La probabilidad de que en un minuto dado, ingresen 50 autos es de aproximadamente: $0.05\%$.

## c) Probabilidad de que en 15 minutos, ingresaran 800 autos
El escenario planteado modifica el intervalo de tiempo original de 1 minuto a un periodo de 15 minutos. Por lo tanto, debemos ajustar el parámetro promedio ($\lambda$) multiplicándolo por la cantidad de minutos analizados:
$$\lambda_{15} = 75 \times 15 = 1125$$

De este modo, la nueva variable aleatoria se define como:
$$Y \sim \text{Poisson}(1125)$$

Con el valor de **$\lambda$** modificado se desea calcular:

Es decir: 
En 15 minutos en promedio hay 1125 autos; Por lo tanto el valor de **$\lambda$** paso de ser de **75** a **1125**

Sustituimos $(y=800)$ y $(\lambda=1125)$ en la función de probabilidad de Poisson:

$$\begin{align*}
P(Y=800) &= \frac{e^{-1125}(1125)^{800}}{800!} \\
\end{align*}$$

Aplicamos el *método de logaritmos* para resolverlo de forma analítica.

Transformamos los productos y cocientes en sumas y restas:

$$\ln(P) = \ln(e^{-1125}) + \ln(1125^{800}) - \ln(800!)$$

Aplicando las propiedades de los logaritmos:
$$\ln(P) = -1125 + 800 \cdot \ln(1125) - \ln(800!)$$

Se usa la función Log-Gamma ($\ln \Gamma(n+1)$), que calcula logaritmos de factoriales de forma exacta mediante series polinomiales de Lanczos o de Stirling extendidas:
Utilizando la aproximación de Stirling (o una calculadora cientifica):

$$\ln(800!) = 4551.950740875012$$

Se hace la resta de forma normal:

$$\begin{align*}
\ln(P) &= -1125 + 5620.43065169137 - 4551.950740875012 \\[2ex]
\ &= -56.52008918364233
\end{align*}$$

Lo devolvemos a la base decimal 10 dividiendo entre ln(10):

$$\begin{align*}
\log_{10}(P) &= \frac{-56.52008918364233}{2.302585092994046} \\
\\
&= -24.546359570138407
\end{align*}$$
Este exponente $-24.54635957$ se separa matemáticamente en su parte entera y su parte decimal:
$$\begin{align*}
\log_{10}(P) &= -24.546359570138407 -(-25)\\
&=0.453640429861593
\end{align*}$$

Elevamos 10 a la parte decimal para obtener el coeficiente:
$$\text{Coeficiente} = 10^{0.453640429861593} \approx \mathbf{2.84211455}$$

Agregando la base del exponente que separamos ($-25$), el resultado final es:
$$P(Y = 800) = 2.8421 \times 10^{-25}$$

Es decir: 

$$(P(Y = 800)) = 0.00000000000000000000000028421$$

La probabilidad de que ingresen exactamente 800 automóviles en un periodo de 15 minutos, dado un promedio esperado de 1125 autos, es de $2.8421 \times 10^{-25}$. Esto representa un evento con una probabilidad prácticamente nula de ocurrir bajo las condiciones normales del modelo por lo que el porcentaje seria muy cercano a 0.
