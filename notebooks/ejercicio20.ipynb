Deportes. En un gimnasio sólo para mujeres se realizan ejercicios aeróbicos y ejercicios en aparatos. El circuito está constituido por nueve aparatos tonificadores y ocho estaciones para los ejercicios aeróbicos. En un día cualquiera, en hora pico (entre las 19 y 20 horas), llegan en promedio siete mujeres. En ese horario: 

a) ¿Cuál es la probabilidad de que lleguen menos de cinco mujeres? 

b) ¿Cuál es la probabilidad de que estén ocupados todos los aparatos? 

c) Y en un día extraordinario, ¿cuál es la probabilidad de que tanto los aparatos como las estaciones estén ocupados?

la funcion que utilizaremos es:
$$P(X = x) = \frac{e^{-\lambda} \cdot \lambda^x}{x!}$$

Donde lambda es 
$\lambda = 7$

## a) Probabilidad de que lleguen menos de 5 mujeres

Para calcular la probabilidad de que lleguen menos de 5 mujeres, es decir, $P(X < 5)$, sumamos las probabilidades individuales desde $x = 0$ hasta $x = 4$:

Es decir:
$$P(X < 5) = P(X = 0) + P(X = 1) + P(X = 2) + P(X = 3) + P(X = 4)$$

Sustituyendo los valores individuales utilizando la fórmula de Poisson con $\lambda = 7$:
$$P(X < 5) = 0.000912 + 0.006383 + 0.022341 + 0.052129 + 0.091226$$

Por lo tanto:

$$P(X < 5) \approx 0.172991$$

Es decir, aproximadamente:

$$17.30\%$$

## b) Probabilidad de que todos los aparatos estén ocupados

Para que los 9 aparatos estén ocupados, deben llegar 9 o más mujeres ($P(X \ge 9)$). Resolvemos utilizando la regla del complemento restando de 1 la suma de las probabilidades desde $x = 0$  hasta  $x = 8$:

$$P(X \ge 9) = 1 - P(X \le 8)$$
$$= 1 - [P(X = 0) + P(X = 1) + \dots + P(X = 8)]$$
$$= 1 - [0.000912 + 0.006383 + 0.022341 + 0.052129 + 0.091226 + 0.127717 + 0.149003 + 0.149003 + 0.130377]$$
$$= 1 - 0.729091$$
$$= 0.270909$$

Por lo tanto:

$$P(X \ge 9) \approx 0.270909$$

Es decir, aproximadamente:

$$27.09\%$$

## c) Probabilidad de que todos los aparatos y estaciones de aeróbicos estén ocupados

Para que todos los lugares estén llenos, se requiere que lleguen 17 o más mujeres ($P(X \ge 17)$), dado que el gimnasio cuenta con 9 aparatos y 8 estaciones de aeróbicos ($9 + 8 = 17$). 

Para resolverlo de forma viable, aplicamos la regla del complemento restando de 1 la suma de todas las probabilidades individuales acumuladas desde $x = 0$ hasta $x = 16$:

Es decir:
$P(X \ge 17) = 1 - P(X \le 16)$

Por lo tanto:
$P(X \ge 17) = 1 - [P(X = 0) + P(X = 1) + P(X = 2) + \dots + P(X = 16)]$

Sustituyendo el valor decimal exacto de cada una de las 17 probabilidades individuales de la distribución de Poisson ($\lambda = 7 $):

* $P(X = 0) = \frac{e^{-7} \cdot 7^0}{0!} = 0.000912$
* $P(X = 1) = \frac{e^{-7} \cdot 7^1}{1!} = 0.006383$
* $P(X = 2) = \frac{e^{-7} \cdot 7^2}{2!} = 0.022341$
* $P(X = 3) = \frac{e^{-7} \cdot 7^3}{3!} = 0.052129$
* $P(X = 4) = \frac{e^{-7} \cdot 7^4}{4!} = 0.091226$
* $P(X = 5) = \frac{e^{-7} \cdot 7^5}{5!} = 0.127717$
* $P(X = 6) = \frac{e^{-7} \cdot 7^6}{6!} = 0.149003$
* $P(X = 7) = \frac{e^{-7} \cdot 7^7}{7!} = 0.149003$
* $P(X = 8) = \frac{e^{-7} \cdot 7^8}{8!} = 0.130377$
* $P(X = 9) = \frac{e^{-7} \cdot 7^9}{9!} = 0.101405$
* $P(X = 10) = \frac{e^{-7} \cdot 7^{10}}{10!} = 0.070983$
* $P(X = 11) = \frac{e^{-7} \cdot 7^{11}}{11!} = 0.045171$
* $P(X = 12) = \frac{e^{-7} \cdot 7^{12}}{12!} = 0.026350$
* $P(X = 13) = \frac{e^{-7} \cdot 7^{13}}{13!} = 0.014188$
* $P(X = 14) = \frac{e^{-7} \cdot 7^{14}}{14!} = 0.007094$
* $P(X = 15) = \frac{e^{-7} \cdot 7^{15}}{15!} = 0.003311$
* $P(X = 16) = \frac{e^{-7} \cdot 7^{16}}{16!} = 0.001448$

Sumando todos los valores anteriores para obtener la probabilidad acumulada:
$P(X \le 16) = 0.000912 + 0.006383 + 0.022341 + 0.052129 + 0.091226 + 0.127717 + 0.149003 + 0.149003 + 0.130377 + 0.101405 + 0.070983 + 0.045171 + 0.026350 + 0.014188 + 0.007094 + 0.003311 + 0.001448 = 0.999252$

Efectuamos la resta del complemento de la siguiente manera:
$P(X \ge 17) = 1 - 0.999252$

Por lo tanto:

$$P(X \ge 17) \approx 0.000748$$

Es decir, aproximadamente:

$$0.0748\%$$
