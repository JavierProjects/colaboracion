$$X=\mbox{número de autos que cuentan con seguro}$$

$$X∼\mbox{Binomial}(n=18,p=0.47)$$

La función de probabilidad es:

$$P(X=x)=\binom{18}{x}(0.47)^x(0.53)^{18-x} \quad \mbox{con} \quad x=0,1,2,\dots,18$$

```python
import matplotlib.pyplot as plt
from scipy.stats import binom

# Parámetros
n = 18
p = 0.47

# Valores posibles de X
x = list(range(n + 1))

# Distribución binomial
probabilidades = binom.pmf(x, n, p)

# Mostrar tabla en consola
#for k, prob in zip(x, probabilidades):
#    print(f"P(X = {k}) = {prob:.6f}")

# ==========================
# Generar tabla HTML
# ==========================

html = """
<table style="border-collapse: collapse; width: 420px; text-align: center;">
  <thead>
    <tr>
      <th style="border: 1px solid #000; padding: 10px; width: 160px;">x</th>
      <th style="border: 1px solid #000; padding: 10px; width: 260px;">P(X=x)</th>
    </tr>
  </thead>
  <tbody>"""

for k, prob in zip(x, probabilidades):
    html += f"""
    <tr>
      <td style="border: 1px solid #000; padding: 8px;">{k}</td>
      <td style="border: 1px solid #000; padding: 8px;">{prob:.6f}</td>
    </tr>"""

html += """
  </tbody>
</table>
"""

# Mostrar el HTML en la consola
print("\nCódigo HTML:\n")
print(html)
```

```python
# ==========================
# Gráfica
# ==========================

plt.figure(figsize=(8, 5))
plt.bar(x, probabilidades, width=1, edgecolor="w")

plt.title("Distribución binomial: autos con seguro")
plt.xlabel("Número de autos con seguro")
plt.ylabel("Probabilidad")
plt.xticks(x)
plt.grid(axis="y", linestyle="--", alpha=0.6)

plt.show()
```

## Tipo de sesgo

La distribución presenta un ligero sesgo a la derecha, también llamado sesgo positivo.

Esto ocurre porque $p=0.47$, es decir, la probabilidad de éxito es levemente menor que $0.5$. La mayor parte de la probabilidad se concentra en los valores cercanos a 8, mientras que la curva se extiende un poco más hacia los valores más altos.

## b) Número esperado de autos que cuenten con seguro

Para una distribución binomial:

$$E(X)=np$$

Por lo tanto tenemos:

$$
\begin{align*}
E(X) &= np \\
     &= 18(0.47) \\
     &= 8.46
\end{align*}
$$

Por lo tanto, se esperarían:

$$8.46$$ autos que cuenten con seguro (aproximadamente 8 o 9 autos).

## c) Probabilidad de que 13 o más autos tengan seguro

“13 o más” significa:

$$P(X \ge 13)$$

Es decir:

$$P(X \ge 13) = P(X=13) + P(X=14) + P(X=15) + P(X=16) + P(X=17) + P(X=18)$$

Por lo tanto: 
$$
\begin{align*}
P(X \geq 13) &= \sum_{x=13}^{18} \binom{18}{x}(0.47)^x(0.53)^{18-x} \\
            &\approx 0.0213 + 0.0069 + 0.0017 + 0.0003 + 0.0000 + 0.0000 \\
            &\approx 0.0302
\end{align*}
$$

Por lo tanto nos da:

$$P(X \ge 13) \approx 0.0302$$

Es decir, aproximadamente: 
$$3.02\%$$.

<span style="color: blue;">Inseguridad</span>. De acuerdo con las estadísticas de criminalidad, en promedio se roba un auto cada tres minutos. Determina:

a) La distribución de probabilidad y la gráfica correspondiente. Describe de manera breve la distribución.

b) La probabilidad de que en los siguientes tres minutos se roben dos autos.

c) La probabilidad de que en los siguientes tres minutos se roben dos autos o menos.

d) La probabilidad de que se roben seis autos en los siguientes 15 minutos.


Dado que evaluamos la probabilidad de que ocurra un número de eventos durante un intervalo de tiempo continuo con un promedio conocido, utilizamos la **Distribución de Poisson**.

Sea:

$$X=\mbox{numero de autos robados en un intervalo de 3 minutos}$$

El promedio ($\lambda$) para un intervalo de 3 minutos es 1:

$$X \sim \mbox{Poisson}(\lambda=1)$$

## a) Distribución de probabilidad

La función de probabilidad de Poisson es:

$$P(X=x) = \frac{e^{-\lambda}\lambda^x}{x!}$$

Sustituyendo nuestra $\lambda = 1$:

$$P(X=x) = \frac{e^{-1}(1)^x}{x!} \quad \mbox{con} \quad x=0,1,2,3,\dots$$

```python
import matplotlib.pyplot as plt
from scipy.stats import poisson

# Parámetro (promedio de robos en 3 minutos)
lam = 1

# Valores posibles de X (tomaremos hasta 7 para mostrar la caída)
x = list(range(8))

# Distribución de Poisson
probabilidades = poisson.pmf(x, lam)

# ==========================
# Generar tabla HTML
# ==========================

html = """
<table style="border-collapse: collapse; width: 420px; text-align: center;">
  <thead>
    <tr>
      <th style="border: 1px solid #000; padding: 10px; width: 160px;">x</th>
      <th style="border: 1px solid #000; padding: 10px; width: 260px;">P(X=x)</th>
    </tr>
  </thead>
  <tbody>"""

for k, prob in zip(x, probabilidades):
    html += f"""
    <tr>
      <td style="border: 1px solid #000; padding: 8px;">{k}</td>
      <td style="border: 1px solid #000; padding: 8px;">{prob:.6f}</td>
    </tr>"""

html += """
  </tbody>
</table>
"""

# Mostrar el HTML en la consola
print("\nCódigo HTML:\n")
print(html)
```

## Descripción de la distribución

La distribución presenta un claro **sesgo positivo (sesgo hacia la derecha)**. 

La mayor probabilidad se concentra en los valores $X=0$ y $X=1$ (ambos con un $\approx 36.78\%$ de probabilidad). A medida que el número de robos aumenta en ese intervalo de 3 minutos, la probabilidad desciende drástically, generando una "cola" que se alarga hacia la derecha.

## b) Probabilidad de que en los siguientes tres minutos se roben dos autos

Evaluamos exactamente $P(X=2)$ usando $\lambda=1$:

$$
\begin{align*}
P(X=2) &= \frac{e^{-1}(1)^2}{2!} \\
       &= \frac{0.367879 \cdot 1}{2} \\
       &= 0.183939
\end{align*}
$$

Por lo tanto:

$$P(X=2) \approx 0.1839$$

Es decir, una probabilidad de **$18.39\%$**.

## c) Probabilidad de que en los siguientes tres minutos se roben dos autos o menos

"Dos autos o menos" significa:

$$P(X \le 2)$$

Es decir, la suma de las probabilidades de $X=0$, $X=1$ y $X=2$:

$$P(X \le 2) = P(X=0) + P(X=1) + P(X=2)$$

Sustituyendo los valores previamente calculados:
$$
\begin{align*}
P(X \le 2) &= 0.367879 + 0.367879 + 0.183940 \\
            &= 0.919698
\end{align*}
$$

Por lo tanto:

$$P(X \le 2) \approx 0.9197$$

Es decir, una probabilidad de **$91.97\%$**.

## d) Probabilidad de que se roben seis autos en los siguientes 15 minutos

Para este inciso, el **intervalo de tiempo cambia**. Si se roba en promedio 1 auto cada 3 minutos, en 15 minutos esperamos que se roben proporcionalmente más.

Ajustamos nuestra nueva media ($\lambda$):
$$\lambda = 1 \cdot \left(\frac{15}{3}\right) = 5$$

Nuestra nueva distribución es $X \sim \mbox{Poisson}(\lambda=5)$. Evaluamos $P(X=6)$:

$$
\begin{align*}
P(X=6) &= \frac{e^{-5}(5)^6}{6!} \\
       &= \frac{0.006738 \cdot 15625}{720} \\
       &= \frac{105.281}{720} \\
       &\approx 0.1462
\end{align*}
$$

Por lo tanto:

$$P(X=6) \approx 0.1462$$

Es decir, una probabilidad de **$14.62\%$**.
