# <span style="color: #289;">Ejercicio 19</span>
## <span style="color: #2965;">Economía.</span> 
### $$\textbf{ Una cadena de supermercados, cuyo nombre empieza con W, abrió 267 tiendas en México durante 2010. 
Con esta información,}$$
$$\text{a) Determina λ para el número promedio de tiendas abiertas en un día.}$$

$$\text{b) ¿Cuál es la probabilidad de que en un día se abran dos tiendas?}$$

$$\text{c) ¿Cuál es la probabilidad de que en una semana se inauguren seis tiendas?}$$

Para esta actividad usaremos la distribución de Poisson

$$P(X=x)=\frac{e^{-\lambda}\lambda^x}{x!}, \quad \mbox{con} \quad x=0,1,2,\ldots$$

Los datos con los que contamos son:

$$λ=\mbox{tiendas funcionando en el año 2010}$$

$$X=\mbox{los días que tuvo en año 2010}$$

$$\textsf{λ = 256}$$

$$\textsf{X = 365}$$

### $$ \text{a) Determina λ para el número promedio de tiendas abiertas en un día.}$$
$$\textsf{Para resolver este ejercicio inciso vamos a usar la formula que indica debemos multiplicar la tasa promedio del evento por el intervalo evaluador }$$
$$λ=\frac{256}{365}$$
$$λ=0.701369\ldots$$

### $$\text{b) ¿Cuál es la posibilidad de que en un día se abran dos tiendas?}$$ 

$$\textsf{En esta pregunta, vamos a tomar como "x" al número 2 para calcular la probabilidad de la abertura de 2 tiendas al día.}$$

$$P(X=x)=\frac{e^{-0.701369\ldots}(0.701369\ldots)^2}{2!}$$
$$P(X=x)=\frac{(0.495905\ldots)*(0.491919\ldots)}{2}$$
$$P(X=x)= 0.128739\ldots$$

### $$ \text{c) ¿Cuál es la probabilidad de que en una semana se inauguren seis tiendas?}$$
$$\textsf{En este inciso, sustituimos los datos conforme la solicitud: "λ=0.7315*7" (λ es multiplicado por 7 para saber que valor tiene por la semana) y "x=6" (para calcular de que abran 6 tiendas). }$$
$$λ=(0.7315\ldots*7) = 5.1205\ldots$$
$$P(X=x)=\frac{e^{-5.1205}(5.1205)^6}{6!}$$
$$P(X=x)=\frac{(5.9730)*(18024.9564)}{720}$$
$$P(X=x)= 0.49532\ldots$$
