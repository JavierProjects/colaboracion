18.- <span style="color: blue;">Inseguridad</span>. De acuerdo con las estadisticas de criminalidad, en promedio se roba **1 auto por cada 3 minutos**. Determina:

a) La distribucion de probabilidad y la grafica correspondiente. Describe de manera breve la distribucion.  
b) La probabilidad de que en los siguiuentes tres minutos se roben 2 autos.  
c) La probabilidad de que en los siguientes tres minutos se roben dos autos o menos.  
d) la probabilidad de que se roben seis autos en los siguientes 15 minutos.  

Sea:

$$X=\text{numero de autos robados por 3 minutos}$$

se conoce tanto el avlor de almbda y x siendo que 


$$\lambda=3.9$$

$$x=\text{la cantidad de eventos que uno quiere descubrir que ocurran}$$

Por lo tanto:

$$X\sim\operatorname{Poisson}(1)$$

La función de probabilidad de Poisson es:


$$P(X=x)=\frac{e^{-\lambda}\lambda^x}{x!}, \quad \mbox{con} \quad x=0,1,2,\ldots$$

Para este problema:


$$P(X=x)=\frac{e^{-1}(1)^x}{x!}$$

## a) Distribución de probabilidad y gráfica

La distribución de Poisson puede tomar una cantidad infinita de valores:


$$x=0,1,2,3,\ldots$$

Para construir una tabla y una gráfica manejables, mostraremos los valores de $(x=0)$ a $(x=0)$. En este intervalo se concentra más del $(99.97\%)$ de la probabilidad total.

Tabla de distribucion de probabilidad con valores aproximadados:

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson
from IPython.display import HTML, display

# Parámetro de la distribución
lam = 1

# Valores que se mostrarán en la tabla y en la gráfica
x = np.arange(0, 11)

# Probabilidades P(X = x)
probabilidades = poisson.pmf(x, mu=lam)

# Construcción de la tabla HTML
html = """
<table style="
    border-collapse: collapse;
    width: 520px;
    text-align: center;
    font-family: Arial, sans-serif;
">
    <thead>
        <tr>
            <th style="
                border: 1px solid #000;
                padding: 10px;
                width: 200px;
            ">x</th>
            <th style="
                border: 1px solid #000;
                padding: 10px;
                width: 320px;
            ">P(X = x)</th>
        </tr>
    </thead>
    <tbody>
"""

for valor, probabilidad in zip(x, probabilidades):
    html += f"""
        <tr>
            <td style="border: 1px solid #000; padding: 8px;">
                {valor}
            </td>
            <td style="border: 1px solid #000; padding: 8px;">
                {probabilidad:.6f}
            </td>
        </tr>
    """

html += """
    </tbody>
</table>
"""

display(HTML(html))
```

```python
# Gráfica de la distribución
plt.figure(figsize=(9, 5))
plt.bar(x, probabilidades, width=1, edgecolor="w")

plt.title("Distribución de Poisson: autos robados entre 3 minutos")
plt.xlabel("Número de autos robados")
plt.ylabel("Probabilidad")
plt.xticks(x)
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.tight_layout()

plt.show()
```

### Descripción de la distribución

La mayor concentración de probabilidad se encuentra alrededor de **0 y 1 autos robados**.

La distribución presenta **sesgo positivo o sesgo a la derecha**, porque:

- la mayor parte de la probabilidad se concentra en los valores pequeños y cercanos al promedio;
- conforme aumenta el número de habitantes, las probabilidades disminuyen;
- la cola de la distribución se extiende hacia los valores grandes de $(X)$.

El sesgo es extremadamente pronunciado porque el valor de $(\lambda=1)$ no permite que la distribución tenga una forma relativamente cercana a una campana y esto hace que sea asimétrica hacia la derecha.

## b) La probabilidad de que en los tres minutos se roben dos autos

Se desea calcular:


$$P(X=2)$$

Sustituimos $(x=5)$ y $(\lambda=3.9)$ en la función de probabilidad de Poisson:

$$
\begin{align*}
P(X=2)
    &= \frac{e^{-1}(1)^2}{2!} \\
    &= \frac{e^{-1}(1)^2}{2} \\
    &= 0.183931 \\    
\end{align*}
$$

Por lo tanto:


$$P(X=5) = 0.1521925$$

La probabilidad de que una vivienda sea habitada por cinco personas es aproximadamente $18.39\%$.

## c) Probabilidad de que en los siguientes tres minutos se roben dos autos o menos

La expresión “dos autos o menos” significa:

$$P(X≤2)$$

Utilizamos la regla del complemento:

$$
\begin{align*}
P(X≤2)
    &= 1-P(X\leq 3) \\
    &= 1-\left[P(X=0)+P(X=1)+P(X=2)+P(X=3)\right] \\
    &= 1-\left[
        \frac{e^{-1}(1)^0}{0!}
        +\frac{e^{-1}(1)^1}{1!}
        +\frac{e^{-1}(1)^2}{2!}
        \right] \\
    &= 1-\left[
        0.367879+0.367879+0.183940
        \right] \\
    &= 0.9195
\end{align*}
$$

Por lo tanto:

$$P(X≤2) = 0.9195$$

La probabilidad de que en los siguientes tres minutos se roben dos autos o menos es aproximadamente $91.95\%$.

## d) Probabilidad de que se roben seis autos en los siguientes 15 minutos

En 15 minutos, la tasa promedio cambia. Si se roba 1 auto cada 3 minutos, en 15 minutos se esperan:

$$\lambda=\frac{15}{3}=5$$

$$
\begin{align*}
P(X=6)
    &= \frac{e^{-5}(5)^6}{6!} \\
    &= \frac{e^{-5}(5)^6}{720} \\
    &= 0.1462 \\
\end{align*}
$$

Por lo tanto:

$$P(X=6) = 0.9195$$

La probabilidad de que se roben seis autos en los siguientes 15 minutos $14.62\%$.

Se quiere realizar una encuesta a 50 personas sobre la calidad de los servicios de salud. Encuentra   

a) ¿Que proporcion de las personas esperas que asistan a los servicios de salud publicos?    
b) ¿Que probabilidad existe de que 30 personas utilicen los servicios que brinda el gobierno   
c) ¿Que probabilidad hay de que cuando mucho 10 utilicen estos servicios?  

Sea:

$$X=\mbox{numero de personas que utilicen los servicios que brinda el gobierno}$$

Como hay 50 encuestas y cada uno tiene probabilidad p=0.34 de utilizar estos servicios:

$$X∼\mbox{Binomial}(n=50,p=0.34)$$

##  a) ¿Que proporcion de las personas esperas que asistan a los servicios de salud publicos? 

La función para conseguir el valor esperado es:

$$\mu = {(n)}{(p)}$$

asi que nuestro valor esperado seria 

$$\mu =  (50)(0.34) = 17$$

se espera que **17 personas** utilicen los servicios de salud

## b) que probabilidad existe de que 30 personas utilicen los servicios que brinda el gobierno?

$$
\begin{align*}
P(X=30) &= \binom{50}{30}(0.34)^30(0.66)^{50-30} \\
       &= \binom{50}{30}(0.34)^30(0.66)^20 \\
       &= 0.000101
\end{align*}
$$

Por lo tanto:

$$P(X=30)\approx 0.000101$$

Es decir, aproximadamente:

$$0.01\%$$
