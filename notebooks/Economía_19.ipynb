## <span style="color: #2965;">Servicios.</span> 
### $$\textbf{ De acuerdo con los datos proporcionados por la Secretaría de Salud en 2010, 34% 
de la población utiliza los servicios de salud públicos federales o estatales. Se quiere realizar una 
encuesta a 50 personas sobre la calidad de los servicios de salud. Encuentra:}$$
$$\text{a) ¿Qué proporción de las personas esperas que asistan a los servicios de salud públicos?}$$

$$\text{b) ¿Qué probabilidad existe de que 30 personas utilicen los servicios que brinda el gobierno?}$$

$$\text{c) ¿Qué probabilidad hay de que cuando mucho 10 utilicen estos servicios?}$$

Para esta actividad usaremos la distribución Binomial

$$P(X=x)=\binom{n}{x}(P)^x(Q)^{n-x}$$

Los datos con los que contamos son:

$$n=\mbox{num. de ensayos de Bernoulli}$$

$$x=\mbox{numero de éxitos}$$

$$P=\mbox{la probabilidad de éxitos}$$

$$Q=\mbox{la probabilidad de fracaso}$$

$$\textsf{n = 50}$$
$$\textsf{x = 30}$$
$$\textsf{P = 0.34}$$
$$\textsf{Q = 0.66}$$

### $$ \text{a) ¿Qué proporción de las personas esperas que asistan a los servicios de salud públicos?}$$
$$\textsf{Para resolver este inciso, vamos a multiplicar la probabilidad de éxitos y la variable fija que nos proporcionan.}$$
$$P(X=x)= N * P $$
$$P(0X=x)= (50) * (0.34) $$
$$P(0X=x)= 17 $$

### $$ \text{b) ¿Qué probabilidad existe de que 30 personas utilicen los servicios que brinda el gobierno?}$$
$$\textsf{Para resolver este inciso, vamos a usar la formula binomial, para saber cuantos usan los servicios del gobierno.}$$
$$P(X=30)=\binom{50}{30}(0.34)^{30}(0.66)^{50-30}$$
$$P(X=30)=\binom{50}{30}(0.34)^{30}(0.66)^{20}$$
$$P(X=30)=(4.71292\ldots)*(8.79766\ldots)*(2.45968\ldots)$$
$$P(X=30)=1.019850241*10^{-4}$$
$$\textsf{(Por temas de precisión yo multiplique mi resultado *100 para obtener el porcentaje.)}$$
$$P(X=30)=(1.019850241*10^{-4}) * (100)$$
$$P(X=30)=0.01019850241$$

### $$ \text{c) ¿Qué probabilidad hay de que cuando mucho 10 utilicen estos servicios?}$$

$$\textsf{En este inciso vamos a resolverlo sumando la distribución binomial desde 0 a 9.}$$

$$P(X=0)=\binom{50}{0}(0.34)^{0}(0.66)^{50} = 2.4*10^{-8}$$
$$P(X=1)=\binom{50}{1}(0.34)^{0}(0.66)^{49} = 3*10^{-7}$$
$$P(X=2)=\binom{50}{2}(0.34)^{0}(0.66)^{48} = 2.5*10^{-6}$$
$$P(X=3)=\binom{50}{3}(0.34)^{0}(0.66)^{47} = 1.5*10^{-5}$$
$$P(X=4)=\binom{50}{4}(0.34)^{0}(0.66)^{46} = 7.2*10^{-5}$$
$$P(X=5)=\binom{50}{5}(0.34)^{0}(0.66)^{45} = 2.8*10^{-4}$$
$$P(X=6)=\binom{50}{6}(0.34)^{0}(0.66)^{44} = 9.1*10^{-4}$$
$$P(X=7)=\binom{50}{7}(0.34)^{0}(0.66)^{43} = 2.5*10^{-3}$$
$$P(X=8)=\binom{50}{8}(0.34)^{0}(0.66)^{42} = 6*10^{-3}$$
$$P(X=9)=\binom{50}{9}(0.34)^{0}(0.66)^{41} = 0.01282960731$$

$$\sum_{i=0}^{10} = 0.01282960731$$
